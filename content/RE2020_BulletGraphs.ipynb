### Graphiques RE2020 en 5 minutes

L'outil produit des graphiques des résultats RE2020 (Bbio, CEP, CEPnr, DH, ICénergie, ICconstruction) : un **bilan bâtiment** en anneaux, un **graphique de conformité RE2020** en bullet graphs, des vues par zone et par indicateur, prêts à coller dans une note ou un rapport. Au démarrage, il est **pré-rempli avec un projet fictif** (2 bâtiments, une zone tertiaire et une zone commerce par bâtiment, un groupe climatisé et un groupe non climatisé par zone) pour montrer le rendu : il faut le remplacer par les résultats du projet.

> ⏳ **Premier chargement** : l'outil télécharge Python dans votre navigateur, cela peut prendre un moment. Les ouvertures suivantes sont plus rapides.

L'interface suit **4 étapes** (boutons en haut, ou « Suivant » en bas de chaque étape).

#### 1 · Projet
- **Projet vierge**, **Projet exemple**, ou **Choisir un JSON** puis **Ouvrir** pour reprendre un projet. **Sauvegarder** : un bouton de téléchargement du `.json` apparaît en bas de l'outil.
- **Structure** : l'arbre **🏢 bâtiments › 📐 zones**. Cliquer sur un bâtiment ou une zone pour le sélectionner ; boutons **+/−** pour ajouter ou supprimer. À droite : nom du bâtiment, nom et surface de la zone, bâtiment de rattachement.
- **Titres des graphiques** : titre, sous-titre, infos en bas.

#### 2 · Saisie
- **A. Importer depuis Excel (recommandé)** : télécharger le modèle, remplir les cellules jaunes (une colonne par zone, ligne `batiment_widget` = bâtiment de la zone, `surface_widget` = surface), puis **Choisir un Excel** et **Importer le fichier**. Le copier-coller et le renvoi des valeurs vers Excel sont dans le volet replié.
- **B. Saisies à corriger** : chaque erreur est un bouton qui ouvre directement la bonne zone et le bon onglet.
- **C. Saisie manuelle** (repliée) : zone en cours, résultats au niveau bâtiment (calculés depuis les zones ou saisis directement depuis Pléiades), puis un onglet par indicateur. Le DH se saisit par groupe (jusqu'à 8 groupes par zone, chacun avec son DHmax et sa cible).

#### 3 · Graphiques
- Cocher les familles à afficher : **Synthèse** (bilan et conformité du projet), **Bâtiment(s)** (bilan et conformité de chaque bâtiment, si plusieurs), **Zone(s)** (vue d'ensemble par zone), **Indicateurs** (un indicateur comparé entre zones), **Individuelles** (une figure par zone et par indicateur). Le nombre de figures est affiché sur chaque bouton.
- **Seules les familles cochées sont calculées.** Après une modification, le bandeau jaune « Modifications non appliquées » s'affiche : cliquer sur **Actualiser**.
- Préréglages : **Note de synthèse** (Synthèse seule) ou **Rapport complet** (tout).
- Aperçu : un graphique à la fois, liste déroulante ou flèches ◀ ▶.
- **Réglages avancés** (replié) : charte WSP ou style de base, largeur, hauteurs, taille des textes, décimales.

#### 4 · Export
- Par défaut : **les graphiques cochés à l'étape 3**. Possible aussi : tous les graphiques, ou une seule famille. Les familles non calculées en aperçu sont calculées à ce moment-là, en haute résolution.
- Cocher **PNG / SVG / PDF**, puis **Exporter (ZIP)**, puis cliquer sur le **bouton de téléchargement**. Si rien ne se passe : clic droit, « Enregistrer le lien sous… ».

> 🔒 **Tout est calculé dans votre navigateur** : aucune donnée de projet n'est envoyée sur un serveur. En contrepartie, rien n'est conservé : téléchargez vos exports et votre JSON avant de fermer ou de recharger la page.

#### Bilan bâtiment et conformité RE2020
- Bbio, CEP, CEPnr, ICénergie et ICconstruction sont vérifiés **au niveau du bâtiment**, le DH **par groupe**.
- **Bilan** : un anneau par indicateur bâtiment et par groupe DH (écart au seuil max en %, valeur / seuil max).
- **Conformité RE2020** : même organisation, avec les bullet graphs (contributions, seuil max, cible, écart au max en rouge si dépassement). Bâtiment *calculé depuis les zones* : contributions de la zone, ou moyenne pondérée par la surface. Bâtiment *saisi directement* : une seule barre.
- La mention **« Le projet est conforme à la RE2020 »** n'apparaît que si **tous** les indicateurs de **tous** les bâtiments sont renseignés et sous leur seuil. Sinon, le message d'état indique ce qui manque ou ce qui dépasse.

#### Bon à savoir
- **Contributions** : chaque indicateur a toujours toutes les contributions du modèle Excel ; une contribution nulle n'est pas tracée. Seul le DH se règle en nombre de groupes.
- **Cible** : facultative, tracée en gris (style de base) ou en pointillé rouge (charte WSP).
- **Couleurs** : logique thermique par défaut. Une couleur modifiée à la main est conservée quand on change de style.

```python
# ============================================================
# 2. STRUCTURE DES DONNÉES
# ============================================================
# - "état brut" (dict JSON) : ce que l'utilisateur saisit, y compris les
#   contributions inactives et les champs encore invalides ;
#   état = {"settings": {...},
#           "buildings": [{"name": ..., "mode": "auto"|"manual", "results": {clé: {"value", "max"}}}],
#           "zones": [{"name": ..., "building": index, "surface": ..., "indicators": {clé: brut}}]}
#   Hiérarchie RE2020 : projet -> bâtiment(s) -> zone(s) -> groupe(s) (DH).
#   Bbio, CEP, CEPnr, ICénergie, ICconstruction : vérifiés au niveau bâtiment ;
#   DH : vérifié par groupe.
# - objets validés (Indicator / Contribution / Settings) : ce qui est tracé.

@dataclass
class Contribution:
    label: str
    value: float
    color: str
    target: float = None          # DH : valeur maximale propre au groupe
    goal: float = None            # DH : cible propre au groupe (facultative)


@dataclass
class Indicator:
    key: str
    unit: str
    target: float
    contributions: list          # uniquement les contributions actives
    include: bool = True
    goal: float = None            # cible facultative (repère discret, en plus du max)
    show_legend: bool = True      # False pour les blocs DH par groupe (le nom est dans le titre)
    heading: str = ""            # titre du bloc (défaut : nom de l'indicateur)
    sub: str = ""                # texte gris à droite du titre (défaut : unité)
    tag: str = ""                # écart au seuil max affiché après le titre (graphique de conformité)
    tag_ko: bool = False          # True = dépassement du seuil max

    @property
    def total(self):
        return float(sum(c.value for c in self.contributions))


@dataclass
class Settings:
    title: str = "Résultats RE2020"
    subtitle: str = ""
    footer: str = ""
    width_cm: float = 20.0
    height_overview_cm: float = 0.0   # 0 = automatique
    height_single_cm: float = 0.0     # 0 = automatique
    font_size: int = 9
    show_values: bool = True
    hide_zero_legend: bool = True     # masque les contributions nulles dans la légende
    common_scale: bool = True         # échelle commune dans les vues « par indicateur »
    decimals: int = 1
    background: str = "blanc"         # "blanc" ou "transparent"
    decimal_sep: str = ","
    dpi: int = 300
    export_dir: str = "exports_RE2020"
    file_prefix: str = "RE2020"
    brand: bool = True                # True = charte WSP, False = style de base
    formats: list = field(default_factory=lambda: ["png", "svg", "pdf"])

    @property
    def transparent(self):
        return self.background == "transparent"


# ---------- Correspondance avec le modèle Excel (clé | valeur)
# Clés normalisées : minuscules, sans suffixe « _widget ».
ICC_LOTS = [
    "Lot 1 – VRD",
    "Lot 2 – Fondations et infrastructure",
    "Lot 3 – Superstructure, maçonnerie",
    "Lot 4 – Couverture, étanchéité, charpente, zinguerie",
    "Lot 5 – Cloisonnement, doublage, plafonds suspendus, menuiseries intérieures",
    "Lot 6 – Façades et menuiseries extérieures",
    "Lot 7 – Revêtements des sols, murs et plafonds, chape, peintures, produits de décoration",
    "Lot 8 – CVC",
    "Lot 9 – Installations sanitaires",
    "Lot 10 – Réseaux d'énergie (courant fort)",
    "Lot 11 – Réseaux de communication (courant faible)",
    "Lot 12 – Appareils élévateurs et autres équipements de transport intérieur",
    "Lot 13 – Équipement de production locale d'électricité",
]

EXCEL_MAP = {
    "Bbio": {"goal": "bbiocible", "show": "show_bbio", "max": "bbiomax", "contribs": [
        ("bbiochaud", "Besoins de chauffage"),
        ("bbiofroid", "Besoins de refroidissement"),
        ("bbioecl", "Besoins d'éclairage")]},
    "CEP": {"goal": "cepcible", "show": "show_cep", "max": "cepmax", "contribs": [
        ("cepch", "Chauffage"), ("cepfr", "Refroidissement"), ("cepecs", "ECS"),
        ("cepecl", "Éclairage"), ("cepauxv", "Auxiliaires de ventilation"),
        ("cepauxd", "Auxiliaires de distribution"), ("cepdep", "Déplacements")]},
    "CEPnr": {"goal": "cepnrcible", "show": "show_cepnr", "max": "cepnrmax", "contribs": [
        ("cepnrch", "Chauffage"), ("cepnrfr", "Refroidissement"), ("cepnrecs", "ECS"),
        ("cepnrecl", "Éclairage"), ("cepnrauxv", "Auxiliaires de ventilation"),
        ("cepnrauxd", "Auxiliaires de distribution"), ("cepnrdep", "Déplacements")]},
    "DH": {"goal": "dhcible", "show": "show_dh", "max": "dhmax", "contribs": [
        (f"dh_g{i}", f"Groupe {i}") for i in range(1, 9)]},
    "ICénergie": {"goal": "icecible", "show": "show_ice", "max": "icemax", "contribs": [
        ("iceelec", "Électricité"), ("icebois", "Bois"), ("icercu", "Réseau de chaleur urbain"),
        ("icefioul", "Fioul"), ("icegaz", "Gaz")]},
    "ICconstruction": {"goal": "icccible", "show": "show_icc", "max": "iccmax", "contribs":
        [(f"icc_lot{i + 1:02d}", lab) for i, lab in enumerate(ICC_LOTS)]
        + [("icc_chantier", "Chantier")]},
}

# Nombre de contributions fixe = structure du modèle (DH : 8 groupes au plus)
for _k, _spec in EXCEL_MAP.items():
    if _k != "DH":
        MAX_CONTRIB[_k] = len(_spec["contribs"])

# Orthographe affichée des clés (identique au modèle Excel)
EXCEL_KEY_NAMES = {k.lower(): k + "_widget" for k in [
    "titre", "soustitre", "infos", "zone", "batiment", "surface", "cepnr_equals_cep",
    "bat_Bbio", "bat_Bbiomax", "bat_CEP", "bat_CEPmax", "bat_CEPnr", "bat_CEPnrmax",
    "bat_IcE", "bat_IcEmax", "bat_IcC", "bat_IcCmax",
    "BbioChaud", "BbioFroid", "BbioEcl", "Bbiomax", "show_bbio",
    "CEPch", "CEPfr", "CEPecs", "CEPecl", "CEPauxV", "CEPauxD", "CEPdep", "CEPmax", "show_cep",
    "CEPnrch", "CEPnrfr", "CEPnrecs", "CEPnrecl", "CEPnrauxV", "CEPnrauxD", "CEPnrdep", "CEPnrmax",
    "show_cepnr", *[f"DH_G{i}" for i in range(1, 9)], *[f"DH_G{i}_nom" for i in range(1, 9)], *[f"DH_G{i}_max" for i in range(1, 9)], "DHmax", "show_dh",
    *[f"DH_G{i}_cible" for i in range(1, 9)], "Bbiocible", "CEPcible", "CEPnrcible", "DHcible",
    "IcEcible", "IcCcible",
    "IcEelec", "IcEbois", "IcErcu", "IcEfioul", "IcEgaz", "IcEmax", "show_ice",
    *[f"IcC_lot{i:02d}" for i in range(1, 14)], "IcC_chantier", "IcCmax", "show_icc"]}


def excel_name(k):
    return EXCEL_KEY_NAMES.get(k, f"{k}_widget")


# ---------- Couleurs par défaut des contributions, logique thermique :
# chauffage = chaud (rouge), froid = bleu, ECS = orange, éclairage = jaune,
# ventilation = vert, distribution = bleu foncé / violet, déplacements = brun.
_BASE = {"ch": "#C8553D", "fr": "#3A7CA5", "ecs": "#E07B39", "ecl": "#E9C46A",
         "auxv": "#81B29A", "auxd": "#5B5F97", "dep": "#9C6644"}
_WSPC = {"ch": "#C33900", "fr": "#1594FD", "ecs": "#F35B05", "ecl": "#FFD10A",
         "auxv": "#496C1A", "auxd": "#162E76", "dep": "#B16F4F"}


def _usage_colors(pal, prefix):
    return {f"{prefix}{k}": pal[k] for k in ("ch", "fr", "ecs", "ecl", "auxv", "auxd", "dep")}


THEME_COLORS = {"base": {}, "wsp": {}}
for _t, _p in (("base", _BASE), ("wsp", _WSPC)):
    THEME_COLORS[_t].update({"bbiochaud": _p["ch"], "bbiofroid": _p["fr"], "bbioecl": _p["ecl"]})
    THEME_COLORS[_t].update(_usage_colors(_p, "cep"))
    THEME_COLORS[_t].update(_usage_colors(_p, "cepnr"))
THEME_COLORS["base"].update({"iceelec": "#E9C46A", "icebois": "#9C6644", "icercu": "#C8553D",
                             "icefioul": "#2F4858", "icegaz": "#6FB7C9"})
THEME_COLORS["wsp"].update({"iceelec": "#FEA513", "icebois": "#B16F4F", "icercu": "#C33900",
                            "icefioul": "#4C2C00", "icegaz": "#162E76"})
_ICC_WSP = ["#404040", "#4C2C00", "#B16F4F", "#496C1A", "#CCCCCC", "#162E76", "#FEA513",
            "#F35B05", "#1594FD", "#FFD10A", "#224807", "#C33900", "#FFE680", "#8C8C8C"]
_DH_WSP = ["#1594FD", "#F35B05", "#496C1A", "#FEA513", "#162E76", "#B16F4F", "#C33900", "#404040"]
for _i, (_ck, _) in enumerate(EXCEL_MAP["ICconstruction"]["contribs"]):
    THEME_COLORS["base"][_ck] = PALETTE[_i % len(PALETTE)]
    THEME_COLORS["wsp"][_ck] = _ICC_WSP[_i % len(_ICC_WSP)]
for _i, (_ck, _) in enumerate(EXCEL_MAP["DH"]["contribs"]):
    THEME_COLORS["base"][_ck] = PALETTE[_i % len(PALETTE)]
    THEME_COLORS["wsp"][_ck] = _DH_WSP[_i % len(_DH_WSP)]


def _spec_color(key, ck, brand=False):
    """Couleur par défaut d'une contribution du modèle (selon le style choisi)."""
    t = THEME_COLORS["wsp" if brand else "base"]
    if ck in t:
        return t[ck]
    cks = [c[0] for c in EXCEL_MAP[key]["contribs"]]
    return PALETTE[cks.index(ck) % len(PALETTE)] if ck in cks else PALETTE[0]


def _default_colors(key, ck, i):
    """Couleurs considérées comme « par défaut » (anciennes versions incluses)."""
    out = {THEME_COLORS["base"].get(ck, "").lower(), THEME_COLORS["wsp"].get(ck, "").lower()}
    cks = [c[0] for c in EXCEL_MAP[key]["contribs"]]
    if ck in cks:
        out.add(PALETTE[cks.index(ck) % len(PALETTE)].lower())
    out.add(PALETTE[i % len(PALETTE)].lower())
    out.discard("")
    return out


def apply_theme_colors(state, brand):
    """Remplace les couleurs restées « par défaut » par celles du style choisi.
    Les couleurs personnalisées (choisies à la main) sont conservées."""
    for z in state["zones"]:
        for key, raw in z["indicators"].items():
            for i, c in enumerate(raw["contributions"]):
                ck = c.get("key") or ""
                if ck and str(c.get("color", "")).lower() in _default_colors(key, ck, i):
                    c["color"] = _spec_color(key, ck, brand)
    return state


def blank_contrib(i):
    return {"key": "", "label": "", "value": "0", "max": "", "goal": "", "color": PALETTE[i % len(PALETTE)]}


def _pad(contribs, key):
    contribs = list(contribs)[:max_contrib(key)]
    return contribs + [blank_contrib(i) for i in range(len(contribs), max_contrib(key))]


def canonical_contribs(key, contribs):
    """Indicateurs hors DH : liste complète des contributions du modèle, dans l'ordre du modèle.
    Les valeurs, libellés et couleurs existants sont repris par clé ; une contribution absente
    vaut 0 (elle n'est pas tracée). Anciens projets sans clés : association par position."""
    spec = EXCEL_MAP[key]["contribs"]
    contribs = [c for c in contribs if isinstance(c, dict)]
    by_key = {c.get("key"): c for c in contribs if c.get("key")}
    if not by_key and contribs:
        by_key = {ck: c for (ck, _), c in zip(spec, contribs)}
    out = []
    for ck, lab in spec:
        c = by_key.get(ck, {})
        out.append({"key": ck, "label": str(c.get("label") or lab),
                    "value": str(c.get("value", "0") if str(c.get("value", "")).strip() else "0"),
                    "max": "", "goal": "", "color": str(c.get("color") or _spec_color(key, ck))})
    return out


def template_raw_indicator(key):
    """Indicateur vierge pré-rempli avec les contributions du modèle Excel (valeurs à 0)."""
    contribs = [{"key": ck, "label": lab, "value": "0", "max": "", "goal": "", "color": _spec_color(key, ck)}
                for ck, lab in EXCEL_MAP[key]["contribs"]]
    n = 1 if key == "DH" else len(contribs)   # DH : un seul groupe actif par défaut
    return {"unit": DEFAULT_UNITS[key], "target": "", "goal": "", "n_active": n,
            "include": True, "contributions": _pad(contribs, key)}


def new_zone(name, building=0):
    return {"name": name, "building": int(building), "surface": "",
            "indicators": {k: template_raw_indicator(k) for k in INDICATOR_KEYS}}


def new_building(name):
    """Bâtiment. mode "auto" : résultats bâtiment déduits des zones (zone unique : valeurs
    de la zone ; plusieurs zones : moyenne pondérée par la surface). mode "manual" : résultats
    bâtiment saisis directement (valeurs Pléiades au niveau bâtiment, recommandé)."""
    return {"name": name, "mode": "auto",
            "results": {k: {"value": "", "max": ""} for k in BUILDING_KEYS}}


def prune_empty_buildings(state):
    """Retire les bâtiments sans zone et sans résultat saisi (au moins un bâtiment conservé).
    Réindexe les zones. Retourne les noms retirés."""
    used = {int(z.get("building", 0)) for z in state["zones"]}
    keep, removed = [], []
    for i, b in enumerate(state["buildings"]):
        has_res = b.get("mode") == "manual" and any(str(r.get("value", "")).strip()
                                                    for r in b["results"].values())
        if i in used or has_res:
            keep.append(i)
        else:
            removed.append(b["name"])
    if not keep:
        keep, removed = [0], removed[1:]
    remap = {old: new for new, old in enumerate(keep)}
    state["buildings"] = [state["buildings"][i] for i in keep]
    for z in state["zones"]:
        z["building"] = remap.get(int(z.get("building", 0)), 0)
    if "current_building" in state:
        state["current_building"] = remap.get(state.get("current_building", 0), 0)
    return removed


def default_state():
    return {"version": 3, "settings": asdict(Settings()), "buildings": [new_building("Bâtiment 1")],
            "zones": [new_zone("Zone 1")], "current_zone": 0, "current_building": 0}


def _num_to_str(v):
    if isinstance(v, bool) or v is None:
        return ""
    if isinstance(v, (int, float)):
        return f"{v:g}" if abs(v) < 1e15 else str(v)
    return str(v)


def raw_from_values(key, target, values, include=True, maxes=None, labels_in=None, goal="", goals=None):
    """Indicateur brut à partir d'un dict {clé Excel: valeur} (ordre du modèle).
    maxes / labels_in : {clé: valeur max / libellé} par contribution (utilisé pour les groupes DH)."""
    labels = dict(EXCEL_MAP[key]["contribs"])
    labels.update(labels_in or {})
    maxes = maxes or {}
    goals = goals or {}
    contribs = [{"key": ck, "label": labels[ck], "value": _num_to_str(values.get(ck, 0)),
                 "max": _num_to_str(maxes.get(ck, "")), "goal": _num_to_str(goals.get(ck, "")),
                 "color": _spec_color(key, ck)}
                for ck, _ in EXCEL_MAP[key]["contribs"] if key != "DH" or ck in values]
    return {"unit": DEFAULT_UNITS[key], "target": _num_to_str(target), "goal": _num_to_str(goal),
            "n_active": max(1, len(contribs)), "include": include,
            "contributions": _pad(contribs or [blank_contrib(0)], key)}


# ---------- Projet de démonstration (FICTIF) : valeurs plausibles RE2020, à remplacer.
# 2 bâtiments × 2 zones (tertiaire / commerce) × 2 groupes DH (climatisé / non climatisé).
# Ordres de grandeur d'un projet de bureaux avec commerces en RDC, seuils 2025 indicatifs :
# ils ne remplacent pas les seuils réels calculés par le logiciel (modulations Mbgéo, Mbcombles,
# Mbsurf, Micat…). Tous les indicateurs sont sous leur seuil : la mention de conformité s'affiche.

def _zone_demo(bbio, cep, cepnr, ice, icc, dh, maxes, goals=None):
    """Indicateurs d'une zone de démonstration.
    bbio/cep/cepnr/ice : {clé: valeur}, icc : liste de 14 valeurs (13 lots + chantier),
    dh : [(nom du groupe, DH, DHmax)], maxes : {indicateur: seuil max}, goals : {indicateur: cible}."""
    goals = goals or {}
    icc_keys = [ck for ck, _ in EXCEL_MAP["ICconstruction"]["contribs"]]
    inds = {
        "Bbio": raw_from_values("Bbio", maxes["Bbio"], bbio, goal=goals.get("Bbio", "")),
        "CEP": raw_from_values("CEP", maxes["CEP"], cep, goal=goals.get("CEP", "")),
        "CEPnr": raw_from_values("CEPnr", maxes["CEPnr"], cepnr, goal=goals.get("CEPnr", "")),
        "ICénergie": raw_from_values("ICénergie", maxes["ICénergie"], ice, goal=goals.get("ICénergie", "")),
        "ICconstruction": raw_from_values("ICconstruction", maxes["ICconstruction"], dict(zip(icc_keys, icc)),
                                          goal=goals.get("ICconstruction", "")),
    }
    gk = [f"dh_g{i + 1}" for i in range(len(dh))]
    inds["DH"] = raw_from_values("DH", "", {k: v for k, (_, v, _) in zip(gk, dh)},
                                 maxes={k: m for k, (_, _, m) in zip(gk, dh)},
                                 labels_in={k: n for k, (n, _, _) in zip(gk, dh)})
    return inds


# Seuils max indicatifs par usage (points / kWhep/(m².an) / kg éq. CO2/m²)
_MAX_TERTIAIRE = {"Bbio": 95, "CEP": 85, "CEPnr": 70, "ICénergie": 280, "ICconstruction": 810}
_MAX_COMMERCE = {"Bbio": 120, "CEP": 110, "CEPnr": 90, "ICénergie": 320, "ICconstruction": 870}
_GOAL_TERTIAIRE = {"Bbio": 80, "ICconstruction": 700}          # cibles internes (exemple)


def _demo_tertiaire(f=1.0, dh=(180, 620)):
    """Zone tertiaire (bureaux). f : coefficient pour différencier les bâtiments."""
    r = lambda v: round(v * f, 1)
    return _zone_demo(
        {"bbiochaud": r(30), "bbiofroid": r(26), "bbioecl": r(18)},
        {"cepch": r(14), "cepfr": r(11), "cepecs": r(2.5), "cepecl": r(12), "cepauxv": r(9),
         "cepauxd": r(2.5), "cepdep": r(3)},
        {"cepnrch": r(13), "cepnrfr": r(10), "cepnrecs": r(2.2), "cepnrecl": r(11), "cepnrauxv": r(8),
         "cepnrauxd": r(2.2), "cepnrdep": r(2.8)},
        {"iceelec": r(150), "icercu": r(45)},
        [r(v) for v in (22, 85, 210, 40, 55, 95, 30, 85, 12, 28, 8, 6, 14, 22)],
        [("Bureaux climatisés", dh[0], 1250), ("Bureaux non climatisés", dh[1], 1250)],
        _MAX_TERTIAIRE, _GOAL_TERTIAIRE)


def _demo_commerce(f=1.0, dh=(95, 840)):
    """Zone commerce (RDC)."""
    r = lambda v: round(v * f, 1)
    return _zone_demo(
        {"bbiochaud": r(38), "bbiofroid": r(34), "bbioecl": r(26)},
        {"cepch": r(17), "cepfr": r(16), "cepecs": r(1.5), "cepecl": r(21), "cepauxv": r(13),
         "cepauxd": r(3), "cepdep": r(0)},
        {"cepnrch": r(15), "cepnrfr": r(14), "cepnrecs": r(1.3), "cepnrecl": r(19), "cepnrauxv": r(12),
         "cepnrauxd": r(2.7), "cepnrdep": r(0)},
        {"iceelec": r(205)},
        [r(v) for v in (35, 110, 190, 45, 40, 120, 45, 95, 10, 35, 9, 0, 0, 25)],
        [("Vente climatisée", dh[0], 1250), ("Réserves non climatisées", dh[1], 1250)],
        _MAX_COMMERCE)


def demo_state():
    """Projet FICTIF de démonstration : 2 bâtiments, 2 zones par bâtiment (tertiaire, commerce),
    2 groupes DH par zone (climatisé, non climatisé). Valeurs plausibles RE2020, à remplacer."""
    st = default_state()
    st["settings"]["title"] = "Projet fictif — Îlot Confluence, bureaux et commerces"
    st["settings"]["subtitle"] = "Phase APD — indice A — valeurs de démonstration"
    st["settings"]["footer"] = ("Infos : valeurs FICTIVES de démonstration (ordres de grandeur RE2020 2025), "
                                "à remplacer par les résultats du projet")
    st["buildings"] = [new_building("Bâtiment A"), new_building("Bâtiment B")]
    st["zones"] = [
        {"name": "Tertiaire", "building": 0, "surface": "2400", "indicators": _demo_tertiaire(1.0, (180, 620))},
        {"name": "Commerce", "building": 0, "surface": "650", "indicators": _demo_commerce(1.0, (95, 840))},
        {"name": "Tertiaire", "building": 1, "surface": "1800", "indicators": _demo_tertiaire(1.06, (240, 710))},
        {"name": "Commerce", "building": 1, "surface": "420", "indicators": _demo_commerce(0.95, (60, 910))},
    ]
    return st


def _normalize_raw_indicator(key, raw_in):
    raw = template_raw_indicator(key)
    if not isinstance(raw_in, dict):
        return raw
    raw["unit"] = str(raw_in.get("unit", raw["unit"]) or "")
    raw["target"] = _num_to_str(raw_in.get("target", ""))
    raw["goal"] = _num_to_str(raw_in.get("goal", ""))
    try:
        n = int(raw_in.get("n_active", 1))
    except (TypeError, ValueError):
        n = 1
    raw["n_active"] = min(max(n, 1), max_contrib(key))
    raw["include"] = bool(raw_in.get("include", True))
    contribs = raw_in.get("contributions", [])
    if isinstance(contribs, list):
        out = []
        for i, c in enumerate(contribs[:max_contrib(key)]):
            if not isinstance(c, dict):
                c = {}
            out.append({"key": str(c.get("key", "") or "").strip().lower(),
                        "label": str(c.get("label", "") or ""),
                        "value": _num_to_str(c.get("value", "")),
                        "max": _num_to_str(c.get("max", "")),
                        "goal": _num_to_str(c.get("goal", "")),
                        "color": str(c.get("color", PALETTE[i % len(PALETTE)]) or PALETTE[i % len(PALETTE)])})
        if key == "DH":
            raw["contributions"] = _pad(out, key)
        else:   # contributions actives uniquement, remises dans la structure fixe du modèle
            raw["contributions"] = canonical_contribs(key, out[:raw["n_active"]])
    if key != "DH":
        raw["n_active"] = max_contrib(key)
    return raw


def normalize_state(data):
    """Fusionne un état chargé (JSON) avec l'état par défaut : tolère les clés manquantes
    et relit les projets de l'ancienne version (une seule zone)."""
    if not isinstance(data, dict):
        raise ValueError("le fichier ne contient pas un objet JSON de projet.")
    state = default_state()
    s_in = data.get("settings", {})
    if isinstance(s_in, dict):
        for k in state["settings"]:
            if k in s_in:
                state["settings"][k] = s_in[k]
    b_in = data.get("buildings")
    buildings = []
    if isinstance(b_in, list):
        for i, b in enumerate(b_in):
            if not isinstance(b, dict):
                continue
            nb = new_building(str(b.get("name", "") or "").strip() or f"Bâtiment {i + 1}")
            nb["mode"] = "manual" if b.get("mode") == "manual" else "auto"
            res = b.get("results") if isinstance(b.get("results"), dict) else {}
            for k in BUILDING_KEYS:
                r = res.get(k) if isinstance(res.get(k), dict) else {}
                nb["results"][k] = {"value": _num_to_str(r.get("value", "")),
                                    "max": _num_to_str(r.get("max", ""))}
            buildings.append(nb)
    state["buildings"] = buildings or [new_building("Bâtiment 1")]
    nbld = len(state["buildings"])
    zones_in = data.get("zones")
    if not isinstance(zones_in, list) or not zones_in:
        zones_in = [{"name": "Zone 1", "indicators": data.get("indicators", {})}]
    zones = []
    for i, z in enumerate(zones_in):
        if not isinstance(z, dict):
            continue
        inds = z.get("indicators", {}) if isinstance(z.get("indicators"), dict) else {}
        name = str(z.get("name", "") or "").strip() or f"Zone {i + 1}"
        try:
            bi = int(z.get("building", 0))
        except (TypeError, ValueError):
            bi = 0
        zones.append({"name": name, "building": min(max(bi, 0), nbld - 1),
                      "surface": _num_to_str(z.get("surface", "")),
                      "indicators": {k: (_normalize_raw_indicator(k, inds[k]) if k in inds
                                         else template_raw_indicator(k)) for k in INDICATOR_KEYS}})
    state["zones"] = zones or [new_zone("Zone 1")]
    try:
        cur = int(data.get("current_zone", 0))
    except (TypeError, ValueError):
        cur = 0
    state["current_zone"] = min(max(cur, 0), len(state["zones"]) - 1)
    try:
        curb = int(data.get("current_building", state["zones"][state["current_zone"]]["building"]))
    except (TypeError, ValueError):
        curb = 0
    state["current_building"] = min(max(curb, 0), nbld - 1)
    return state
```

```python
# ============================================================
# 3. VALIDATION
# ============================================================

@dataclass
class ValidationResult:
    indicator: object = None      # Indicator si valide, sinon None
    errors: list = field(default_factory=list)
    warnings: list = field(default_factory=list)

    @property
    def ok(self):
        return not self.errors and self.indicator is not None


def parse_number(text):
    """Convertit une saisie en nombre. Accepte la virgule décimale et les espaces.
    Retourne (valeur, message_erreur)."""
    if isinstance(text, bool):
        return None, "valeur booléenne"
    if isinstance(text, (int, float)):
        return (float(text), None) if math.isfinite(text) else (None, "valeur non finie")
    s = "" if text is None else str(text)
    s = s.strip().replace("\u202f", "").replace("\u00a0", "").replace(" ", "")
    if not s:
        return None, "champ vide"
    s = s.replace(",", ".")
    try:
        v = float(s)
    except ValueError:
        return None, f"« {str(text).strip()} » n'est pas un nombre"
    if not math.isfinite(v):
        return None, "valeur non finie"
    return v, None


def normalize_color(c, fallback):
    try:
        if c and mcolors.is_color_like(c):
            return mcolors.to_hex(c), True
    except (ValueError, TypeError):
        pass
    return fallback, False


def validate_indicator(key, raw):
    res = ValidationResult()
    try:
        n = int(raw.get("n_active", 1))
    except (TypeError, ValueError):
        n = 1
    n = min(max(n, 1), max_contrib(key)) if key == "DH" else max_contrib(key)   # hors DH : toutes

    unit = str(raw.get("unit", "") or "").strip()
    if not unit:
        res.warnings.append("unité non renseignée")

    per_group = key == "DH"   # DH : une valeur maximale par groupe (DHmax = valeur par défaut)
    target, err = parse_number(raw.get("target", ""))
    dh_default = None if err else target
    if per_group:
        g_max = [parse_number(raw["contributions"][i].get("max", "")) for i in range(n)]
        if err and all(e for _, e in g_max):
            res.errors.append("Valeur maximale : renseigner un DHmax par groupe ou un DHmax par défaut.")
        elif not err and target <= 0:
            res.errors.append("DHmax par défaut : il doit être strictement positif.")
        if err:
            ok_vals = [v for v, e in g_max if not e and v > 0]
            target = max(ok_vals) if ok_vals else 0.0
    elif err:
        res.errors.append(f"Valeur maximale : {err}.")
    elif target <= 0:
        res.errors.append("Valeur maximale : elle doit être strictement positive.")

    goal = None
    g, gerr = parse_number(raw.get("goal", ""))
    if gerr is None:
        if g <= 0:
            res.errors.append("Cible : elle doit être strictement positive (ou laisser vide).")
        else:
            goal = g
            if not per_group and not err and target > 0 and g > target:
                res.warnings.append("la cible est supérieure à la valeur maximale.")
    elif str(raw.get("goal", "") or "").strip():
        res.errors.append(f"Cible : {gerr}.")
    dh_goal_default = goal if per_group else None

    contribs = []
    for i in range(n):
        c = raw["contributions"][i]
        label = str(c.get("label", "") or "").strip() or f"Contribution {i + 1}"
        v, err = parse_number(c.get("value", ""))
        if err:
            res.errors.append(f"Contribution {i + 1} ({label}) : {err}.")
            continue
        if v < 0:
            res.errors.append(f"Contribution {i + 1} ({label}) : valeur négative non admise.")
            continue
        color, ok = normalize_color(c.get("color"), PALETTE[i % len(PALETTE)])
        if not ok:
            res.warnings.append(f"Contribution {i + 1} : couleur invalide, couleur par défaut utilisée.")
        c_target = None
        if per_group:
            m, merr = parse_number(c.get("max", ""))
            if merr is None and m <= 0:
                res.errors.append(f"Groupe {i + 1} ({label}) : DHmax doit être strictement positif.")
                continue
            if merr is None:
                c_target = m
            elif dh_default and dh_default > 0:
                c_target = dh_default
            else:
                res.errors.append(f"Groupe {i + 1} ({label}) : DHmax manquant.")
                continue
        c_goal = None
        if per_group:
            gm, gmerr = parse_number(c.get("goal", ""))
            if gmerr is None and gm <= 0:
                res.errors.append(f"Groupe {i + 1} ({label}) : la cible doit être strictement positive.")
                continue
            c_goal = gm if gmerr is None else dh_goal_default
        contribs.append(Contribution(label, v, color, c_target, c_goal))

    if not res.errors:
        res.indicator = Indicator(key, unit, target, contribs, bool(raw.get("include", True)),
                                  goal=None if per_group else goal)
    return res


def validate_settings(raw):
    """Contrôle et borne les paramètres généraux (jamais bloquant)."""
    s = Settings()
    warnings = []

    def _num(name, cast, lo, hi):
        val = raw.get(name, getattr(s, name))
        try:
            val = cast(val)
            if not math.isfinite(float(val)):
                raise ValueError
        except (TypeError, ValueError):
            warnings.append(f"Paramètre « {name} » invalide : valeur par défaut utilisée.")
            return getattr(s, name)
        if val < lo or val > hi:
            warnings.append(f"Paramètre « {name} » ramené dans l'intervalle [{lo} ; {hi}].")
            val = min(max(val, lo), hi)
        return val

    s.title = str(raw.get("title", s.title) or "")
    s.subtitle = str(raw.get("subtitle", s.subtitle) or "")
    s.footer = str(raw.get("footer", s.footer) or "")
    s.width_cm = _num("width_cm", float, 8.0, 60.0)
    s.height_overview_cm = _num("height_overview_cm", float, 0.0, 100.0)
    s.height_single_cm = _num("height_single_cm", float, 0.0, 100.0)
    s.font_size = _num("font_size", int, 6, 20)
    s.decimals = _num("decimals", int, 0, 4)
    s.dpi = _num("dpi", int, 300, 1200)
    s.show_values = bool(raw.get("show_values", True))
    s.hide_zero_legend = bool(raw.get("hide_zero_legend", True))
    s.common_scale = bool(raw.get("common_scale", True))
    s.background = "transparent" if raw.get("background") == "transparent" else "blanc"
    s.decimal_sep = "." if raw.get("decimal_sep") == "." else ","
    s.brand = bool(raw.get("brand", True))
    s.export_dir = str(raw.get("export_dir", s.export_dir) or "").strip() or "exports_RE2020"
    s.file_prefix = str(raw.get("file_prefix", s.file_prefix) or "").strip() or "RE2020"
    fmts = raw.get("formats", s.formats)
    if not isinstance(fmts, (list, tuple)):
        fmts = s.formats
    s.formats = [f for f in ("png", "svg", "pdf") if f in fmts]
    return s, warnings

# ---------- Bilan bâtiment : résultats au niveau bâtiment (Bbio, CEP, CEPnr, ICénergie,
# ICconstruction) et DH au niveau de chaque groupe de chaque zone du bâtiment.

@dataclass
class BilanItem:
    key: str          # indicateur (ou « DH »)
    label: str        # texte affiché au-dessus de l'anneau
    value: float
    max: float
    unit: str

    @property
    def ok(self):
        return self.value <= self.max * (1 + 1e-9)

    @property
    def gap(self):
        """Écart relatif au seuil max : (valeur - max) / max."""
        return (self.value - self.max) / self.max


@dataclass
class BuildingBilan:
    name: str
    items: list = field(default_factory=list)      # BilanItem des indicateurs bâtiment
    dh: list = field(default_factory=list)         # BilanItem par groupe DH
    missing: list = field(default_factory=list)    # indicateurs absents (conformité non démontrée)
    bullets: list = field(default_factory=list)    # Indicator bâtiment (bullet graph de conformité)
    dh_bullets: list = field(default_factory=list) # Indicator par groupe DH
    errors: list = field(default_factory=list)
    warnings: list = field(default_factory=list)

    @property
    def complete(self):
        return not self.missing and not self.errors and bool(self.items) and bool(self.dh)

    @property
    def conform(self):
        """Conforme seulement si TOUS les indicateurs sont présents et sous leur seuil."""
        return self.complete and all(i.ok for i in self.items + self.dh)

    @property
    def exceed(self):
        return [i for i in self.items + self.dh if not i.ok]


def zones_of(state, bi):
    return [z for z in state["zones"] if int(z.get("building", 0)) == bi]


def compute_building_bilan(state, bi):
    """Calcule le bilan d'un bâtiment à partir de l'état brut."""
    b = state["buildings"][bi]
    out = BuildingBilan(b["name"])
    zs = zones_of(state, bi)
    if not zs:
        out.errors.append("aucune zone n'est rattachée à ce bâtiment.")
    for k in BUILDING_KEYS:
        unit = DEFAULT_UNITS[k]
        if b.get("mode") == "manual":
            r = b["results"].get(k, {})
            v, ve = parse_number(r.get("value", ""))
            m, me = parse_number(r.get("max", ""))
            if ve and me and not str(r.get("value", "")).strip() and not str(r.get("max", "")).strip():
                out.missing.append(k)
            elif ve or me:
                out.errors.append(f"{k} : valeur et seuil max à renseigner (saisie bâtiment).")
            elif m <= 0 or v < 0:
                out.errors.append(f"{k} : seuil max strictement positif et valeur positive attendus.")
            else:
                out.items.append(BilanItem(k, k, v, m, unit))
                # pas de détail par contribution en saisie directe : une seule barre (couleur à la tracée)
                out.bullets.append(Indicator(k, unit, m, [Contribution("Résultat bâtiment", v, "")],
                                             show_legend=False))
            continue
        # mode automatique : à partir des zones du bâtiment
        vals, excluded = [], []
        for z in zs:
            raw = z["indicators"][k]
            if not raw["include"]:
                excluded.append(z["name"])
                continue
            res = validate_indicator(k, raw)
            if not res.ok:
                out.errors.append(f"{k} — zone « {z['name']} » : saisie à corriger.")
                continue
            vals.append((z, res.indicator))
            unit = res.indicator.unit or unit
        if not vals:
            if not any(f"{k} —" in e for e in out.errors):
                out.missing.append(k)
            continue
        if excluded:
            out.warnings.append(f"{k} : zone(s) exclue(s) du calcul bâtiment : " + ", ".join(excluded) + ".")
        if len(vals) == 1:
            z, ind = vals[0]
            out.items.append(BilanItem(k, k, ind.total, ind.target, unit))
            out.bullets.append(replace(ind, heading="", sub=""))
            continue
        surf = []
        for z, _ in vals:
            a, ae = parse_number(z.get("surface", ""))
            surf.append(None if ae or a <= 0 else a)
        if any(a is None for a in surf):
            out.errors.append(f"{k} : plusieurs zones dans le bâtiment, renseigner la surface de chaque zone "
                              f"ou saisir directement les résultats bâtiment.")
            continue
        A = sum(surf)
        v = sum(ind.total * a for (_, ind), a in zip(vals, surf)) / A
        m = sum(ind.target * a for (_, ind), a in zip(vals, surf)) / A
        out.items.append(BilanItem(k, k, v, m, unit))
        # contributions pondérées par la surface (structure fixe : même contribution à la même place)
        ref = vals[0][1]
        if any(len(ind.contributions) != len(ref.contributions) for _, ind in vals):
            out.bullets.append(Indicator(k, unit, m, [Contribution("Résultat bâtiment", v, "")],
                                         show_legend=False))
            continue
        contribs = [Contribution(c.label, sum(ind.contributions[i].value * a
                                              for (_, ind), a in zip(vals, surf)) / A, c.color)
                    for i, c in enumerate(ref.contributions)]
        goals = [ind.goal for _, ind in vals]
        goal = (sum(g * a for g, a in zip(goals, surf)) / A) if all(goals) else None
        out.bullets.append(Indicator(k, unit, m, contribs, goal=goal))
    # DH : un anneau par groupe de chaque zone
    multi = len(zs) > 1
    for z in zs:
        raw = z["indicators"]["DH"]
        if not raw["include"]:
            continue
        res = validate_indicator("DH", raw)
        if not res.ok:
            out.errors.append(f"DH — zone « {z['name']} » : saisie à corriger.")
            continue
        ind = res.indicator
        for c in ind.contributions:
            tgt = c.target if c.target else ind.target
            lab = f"{z['name']} — {c.label}" if multi else c.label
            out.dh.append(BilanItem("DH", lab, c.value, tgt, ind.unit or DEFAULT_UNITS["DH"]))
            out.dh_bullets.append(Indicator("DH", ind.unit, tgt, [Contribution(c.label, c.value, c.color, tgt, c.goal)],
                                            True, goal=c.goal, show_legend=False, heading=f"DH — {lab}"))
    if not out.dh and not any(e.startswith("DH") for e in out.errors):
        out.missing.append("DH")
    if out.missing:
        out.warnings.append("Indicateur(s) absent(s) : " + ", ".join(out.missing)
                            + " — conformité RE2020 non affichée.")
    return out


def project_bilan(state):
    """Liste des bilans de tous les bâtiments + conformité globale du projet."""
    bl = [compute_building_bilan(state, i) for i in range(len(state["buildings"]))]
    return bl, bool(bl) and all(b.conform for b in bl)
```

```python
# ============================================================
# 4. FONCTIONS DE TRACÉ (Matplotlib, API objet — aucune figure pyplot globale)
# ============================================================

BAR_HALF = 0.28          # demi-hauteur de la barre (axe y de -0.5 à 0.5)


def fmt_num(v, decimals, sep):
    s = f"{v:,.{int(decimals)}f}"
    if sep == ",":
        s = s.replace(",", "\u00a0").replace(".", ",")
    return s


def fmt_target(v, decimals, sep):
    d = 0 if abs(v - round(v)) < 1e-9 else decimals
    return fmt_num(v, d, sep)


def _font_sizes(base):
    return {"title": base * 1.45, "head": base * 1.12, "ann": base * 0.92,
            "tick": base * 0.85, "leg": base * 0.88, "seg": base * 0.82}


def _text_color_for(bg):
    """Noir ou blanc selon le meilleur contraste (WCAG) avec la couleur du segment."""
    r, g, b = mcolors.to_rgb(bg)
    lin = [c / 12.92 if c <= 0.03928 else ((c + 0.055) / 1.055) ** 2.4 for c in (r, g, b)]
    lum = 0.2126 * lin[0] + 0.7152 * lin[1] + 0.0722 * lin[2]
    return "#1A1A1A" if (lum + 0.05) / 0.05 >= 1.05 / (lum + 0.05) else "#FFFFFF"


def _new_figure(w_in, h_in):
    fig = Figure(figsize=(w_in, h_in), dpi=100)
    FigureCanvasAgg(fig)
    return fig


def _legend_kwargs(fs):
    return dict(frameon=False, fontsize=fs, handlelength=1.1, handleheight=0.9,
                handletextpad=0.5, columnspacing=1.8, labelspacing=0.45,
                borderpad=0.1, borderaxespad=0.0)


def _legend_entries(ind, ncol):
    """Ordonne les entrées pour une lecture ligne par ligne (même ordre que la barre)."""
    handles = [Patch(facecolor=c.color, edgecolor="none") for c in ind.contributions]
    labels = [textwrap.fill(c.label, LEGEND_WRAP) for c in ind.contributions]
    n = len(labels)
    ncol = max(1, min(ncol, n))
    nrows, n_large = divmod(n, ncol)
    rows_per_col = [nrows + 1] * n_large + [nrows] * (ncol - n_large)
    order = [r * ncol + c for c in range(ncol) for r in range(rows_per_col[c])]
    return [handles[i] for i in order], [labels[i] for i in order]


def _measure_legend(ind, avail_w_in, fs):
    """Nombre de colonnes maximal tenant dans la largeur, puis hauteur réelle mesurée."""
    fig = _new_figure(max(avail_w_in * 2, 6), 6)
    r = fig.canvas.get_renderer()
    n = len(ind.contributions)
    if n == 0:
        return 1, 0.0

    def size(ncol):
        h, l = _legend_entries(ind, ncol)
        leg = fig.legend(h, l, ncol=ncol, loc="upper left", **_legend_kwargs(fs))
        bb = leg.get_window_extent(r)
        leg.remove()
        return bb.width / fig.dpi, bb.height / fig.dpi

    best = 1
    for ncol in range(n, 0, -1):
        if size(ncol)[0] <= avail_w_in:
            best = ncol
            break
    rows = math.ceil(n / best)
    best = math.ceil(n / rows)            # colonnes équilibrées pour ce nombre de lignes
    w, h = size(best)
    return best, h


def _block_metrics(ind, s, avail_w):
    fs = _font_sizes(s.font_size)
    if ind.show_legend:
        ncol, leg_h = _measure_legend(ind, avail_w, fs["leg"])
    else:
        ncol, leg_h = 1, 0.0
    m = {
        "ncol": ncol,
        "head_h": fs["head"] * 1.5 / 72,
        "ann_h": fs["ann"] * 1.6 / 72,
        "bar_h": max(0.30, s.font_size * 2.6 / 72),
        "tick_h": fs["tick"] * 2.2 / 72,
        "leg_gap": 0.06 if ind.show_legend else 0.0,
        "leg_h": leg_h,
    }
    m["total"] = sum(m[k] for k in ("head_h", "ann_h", "bar_h", "tick_h", "leg_gap", "leg_h"))
    return m


def _tick_formatter(xmax, sep):
    loc = MaxNLocator(nbins=6, steps=[1, 2, 2.5, 5, 10], min_n_ticks=3)
    ticks = [t for t in loc.tick_values(0, xmax) if 0 <= t <= xmax * 1.0001]
    d = 0
    for d in range(0, 5):
        if all(abs(t * 10 ** d - round(t * 10 ** d)) < 1e-6 for t in ticks):
            break
    return loc, FuncFormatter(lambda v, pos: fmt_num(v, d, sep))


def _text_width_pt(fig, txt, size, weight="normal"):
    """Largeur d'un texte en points (mesurée avec le moteur de rendu de la figure)."""
    r = fig.canvas.get_renderer()
    t = fig.text(0, 0, txt, fontsize=size, fontweight=weight)
    w = t.get_window_extent(r).width / (fig.dpi / 72)
    t.remove()
    return w


def _draw_bullet(fig, ax, ind, s, m, min_xmax=0.0):
    """Trace un bullet graph : barre empilée + repère de cible + annotations."""
    fs = _font_sizes(s.font_size)
    th = theme(s)
    r = fig.canvas.get_renderer()
    px_per_pt = fig.dpi / 72
    total, target = ind.total, ind.target
    goal = ind.goal if ind.goal and ind.goal > 0 else None

    ax.set_ylim(-0.5, 0.5)
    ax.set_yticks([])
    for side in ("top", "right", "left"):
        ax.spines[side].set_visible(False)
    ax.spines["bottom"].set_color(th["axis"])
    ax.spines["bottom"].set_linewidth(0.8)
    ax.set_facecolor("none")
    ax.set_axisbelow(True)
    ax.grid(axis="x", color=th["grid"], linewidth=0.6)

    # Titre du bloc (gras) + complément (gris : unité, zone…)
    head_txt = ind.heading or ind.key
    if th["upper"]:
        head_txt = head_txt.upper()
    dx = 8 if th["head_bar"] else 0        # place du petit trait rouge WSP
    head = ax.annotate(head_txt, xy=(0, 1), xycoords="axes fraction",
                       xytext=(dx, m["ann_h"] * 72), textcoords="offset points",
                       ha="left", va="bottom", fontsize=fs["head"], fontweight="bold",
                       color=th["text"])
    sub = ind.sub if ind.sub else (f"({ind.unit})" if ind.unit else "")
    if sub:
        w_head_pt = head.get_window_extent(r).width / px_per_pt
        ax.annotate(sub, xy=(0, 1), xycoords="axes fraction",
                    xytext=(dx + w_head_pt + 5, m["ann_h"] * 72), textcoords="offset points",
                    ha="left", va="bottom", fontsize=fs["head"] * 0.9, color=th["muted"])
    if ind.tag:   # écart au seuil max (graphique de conformité), après le titre et l'unité
        x_pt = dx + head.get_window_extent(r).width / px_per_pt + 5
        if sub:
            x_pt += _text_width_pt(fig, sub, fs["head"] * 0.9) + 8
        mode = "wsp" if s.brand else "base"
        c_tag = BILAN_STATUS[mode]["ko"] if ind.tag_ko else ("#1E6B3A" if mode == "base" else th["text"])
        ax.annotate(ind.tag, xy=(0, 1), xycoords="axes fraction",
                    xytext=(x_pt, m["ann_h"] * 72), textcoords="offset points",
                    ha="left", va="bottom", fontsize=fs["head"] * 0.9, fontweight="bold", color=c_tag)

    # Segments empilés
    segments, left = [], 0.0
    for c in ind.contributions:
        if c.value > 0:
            ax.barh(0, c.value, left=left, height=2 * BAR_HALF, color=c.color,
                    edgecolor="white", linewidth=0.8, zorder=2)
        segments.append((left, c.value, c))
        left += c.value

    # Repère de cible : trait foncé avec liseré blanc (visible sur tout segment)
    ax.plot([target, target], [-0.42, 0.5], color=th["max"], linewidth=1.8,
            solid_capstyle="butt", zorder=5, clip_on=False,
            path_effects=[patheffects.withStroke(linewidth=3.8, foreground="white")])

    # Repère de cible (facultatif) : plus fin, gris, pointillé
    if goal is not None:
        ax.plot([goal, goal], [-0.36, 0.44], color=th["goal"], linewidth=1.3,
                linestyle=(0, (3, 1.6)) if th["goal_dash"] else "-",
                solid_capstyle="butt", zorder=5, clip_on=False,
                path_effects=[patheffects.withStroke(linewidth=2.4, foreground="white")])

    total_txt = ax.annotate(f"Total : {fmt_num(total, s.decimals, s.decimal_sep)}",
                            xy=(total, 0), xytext=(6, 0), textcoords="offset points",
                            ha="left", va="center", fontsize=fs["ann"], fontweight="bold",
                            color=th["text"], zorder=6, annotation_clip=False)
    tgt_txt = ax.annotate(f"Max : {fmt_target(target, s.decimals, s.decimal_sep)}",
                          xy=(target, 1), xycoords=("data", "axes fraction"),
                          xytext=(0, 2), textcoords="offset points",
                          ha="center", va="bottom", fontsize=fs["ann"],
                          color=th["max"], fontweight="bold" if th["upper"] else "normal",
                          zorder=6, annotation_clip=False)

    # Échelle propre à l'indicateur : 0 -> max(total, cible) + place pour les textes
    M = max(total, target, goal or 0)
    ax.set_xlim(0, M * 1.04)
    w_px = ax.bbox.width
    w_tot = total_txt.get_window_extent(r).width
    w_tgt = tgt_txt.get_window_extent(r).width
    off, pad = 6 * px_per_pt, 4 * px_per_pt

    def _needs_shift(xmax):
        # Si la cible tombe juste après la barre, le total est décalé à droite de la cible
        sc = w_px / xmax
        return total <= target and (target - total) * sc < off + w_tot + off

    def _anchor(xmax):
        # Le total est placé après le dernier repère (cible ou max) qui le chevaucherait
        sc = w_px / xmax
        a = total
        for mk in sorted(x for x in (goal, target) if x is not None and x >= total):
            if (mk - a) * sc < off + w_tot + off:
                a = mk
        return a

    xmax = M * 1.04
    for _ in range(20):
        anchor = _anchor(xmax)
        d1 = max(w_px - off - w_tot - pad, 0.2 * w_px)
        d2 = max(w_px - w_tgt / 2 - pad, 0.2 * w_px)
        new = max(xmax, anchor * w_px / d1, target * w_px / d2)
        if new <= xmax * (1 + 1e-9):
            break
        xmax = new
    xmax = max(xmax, min_xmax)          # échelle commune éventuelle
    anchor = _anchor(xmax)
    total_txt.xy = (anchor, 0)
    ax.set_xlim(0, xmax)
    sc = w_px / xmax

    # Annotation du max : alignée à gauche si elle déborderait à gauche
    if target * sc < w_tgt / 2 + pad:
        tgt_txt.set_ha("left")
        tgt_txt.set_position((-1, 2))

    # Annotation de la cible : du côté opposé au max pour éviter le chevauchement ;
    # si la place manque, elle est regroupée avec le libellé du max.
    if goal is not None:
        g_lbl = f"Cible : {fmt_target(goal, s.decimals, s.decimal_sep)}"
        left_side = goal <= target
        g_txt = ax.annotate(g_lbl, xy=(goal, 1), xycoords=("data", "axes fraction"),
                            xytext=(-3 if left_side else 3, 2), textcoords="offset points",
                            ha="right" if left_side else "left", va="bottom", fontsize=fs["ann"] * 0.95,
                            color=th["goal_label"], zorder=6, annotation_clip=False)
        gb, tb = g_txt.get_window_extent(r), tgt_txt.get_window_extent(r)
        ab = ax.get_window_extent(r)
        overlap = gb.x1 > tb.x0 - 2 and gb.x0 < tb.x1 + 2
        if overlap or gb.x0 < ab.x0 - 2 or gb.x1 > ab.x1 + 2:
            g_txt.remove()
            tgt_txt.set_text(tgt_txt.get_text() + "  |  " + g_lbl)
            tb2 = tgt_txt.get_window_extent(r)
            if tb2.x0 < ab.x0 - 2:
                tgt_txt.set_ha("left")
                tgt_txt.set_position((-1, 2))
            elif tb2.x1 > ab.x1 + 2:
                tgt_txt.set_ha("right")
                tgt_txt.set_position((2, 2))

    # Valeurs dans les segments : seulement si lisibles et non traversées par la cible
    if s.show_values:
        bar_px = ax.bbox.height * 2 * BAR_HALF
        for lft, v, c in segments:
            if v <= 0:
                continue
            t = ax.text(lft + v / 2, 0, fmt_num(v, s.decimals, s.decimal_sep),
                        ha="center", va="center", fontsize=fs["seg"],
                        color=_text_color_for(c.color), zorder=4)
            bb = t.get_window_extent(r)
            fits = bb.width + 2 * pad <= v * sc and bb.height <= bar_px * 0.9
            hits_target = any(abs(x * sc - (lft + v / 2) * sc) < bb.width / 2 + 3 * px_per_pt
                              for x in (target, goal) if x is not None)
            if not fits or hits_target:
                t.remove()

    loc, fmt = _tick_formatter(xmax, s.decimal_sep)
    ax.xaxis.set_major_locator(loc)
    ax.xaxis.set_major_formatter(fmt)
    ax.tick_params(axis="x", labelsize=fs["tick"], length=3, width=0.6,
                   color=th["axis"], labelcolor=th["muted"], pad=2)
    return xmax


def _wrap_title(title, avail_w_in, fontsize):
    """Renvoie le titre découpé en lignes pour tenir dans la largeur disponible."""
    title = title.strip()
    if not title:
        return ""
    fig = _new_figure(max(avail_w_in * 3, 6), 2)
    r = fig.canvas.get_renderer()
    for width in range(len(title), 9, -2):
        wrapped = textwrap.fill(title, width)
        t = fig.text(0, 0.5, wrapped, fontsize=fontsize, fontweight="bold")
        ok = t.get_window_extent(r).width / fig.dpi <= avail_w_in
        t.remove()
        if ok:
            return wrapped
    return textwrap.fill(title, 10)


def _build(indicators, s, height_cm, title, subtitle, zone_name, footer, min_xmax):
    with _rc():
        fs = _font_sizes(s.font_size)
        th = theme(s)
        if th["upper"]:
            title = (title or "").upper()
            zone_name = (zone_name or "").upper()
        Wd = s.width_cm / 2.54
        margin_l, margin_r = 0.30, 0.40
        avail = Wd - margin_l - margin_r
        metrics = [_block_metrics(ind, s, avail) for ind in indicators]
        n = len(indicators)
        top_pad, bottom_pad = 0.12, 0.10
        title = _wrap_title(title, avail, fs["title"])
        subtitle = _wrap_title(subtitle, avail, fs["head"])
        zone_name = _wrap_title(zone_name, avail, fs["head"])
        footer = _wrap_title(footer, avail, fs["ann"])
        title_h = (fs["title"] * (1.2 * (title.count("\n") + 1) + 0.35) / 72) if title else 0.0
        subtitle_h = (fs["head"] * (1.2 * (subtitle.count("\n") + 1) + 0.25) / 72) if subtitle else 0.0
        zone_h = (fs["head"] * 1.7 / 72) if zone_name else 0.0
        footer_gap = 0.25
        footer_h = (footer_gap + fs["ann"] * (1.25 * (footer.count("\n") + 1) + 0.4) / 72) if footer else 0.0
        gap = max(0.20, s.font_size * 2.2 / 72)
        rule_h = 0.12 if th["banner"] else 0.0
        band_pad = (0.06, 0.12) if th["banner"] else (0.0, 0.0)
        after_band = 0.16 if th["banner"] else 0.0
        h_req = rule_h + sum(band_pad) + after_band + top_pad + title_h + subtitle_h + zone_h + sum(m["total"] for m in metrics) + gap * (n - 1) + footer_h + bottom_pad

        H, note = h_req, None
        if height_cm and height_cm > 0:
            h_user = height_cm / 2.54
            if h_user >= h_req:
                H = h_user
                extra = h_user - h_req
                if n > 1:
                    gap += extra / (n - 1)
                else:
                    top_pad += extra / 2
            else:
                note = (f"hauteur demandée ({height_cm:.1f} cm) insuffisante : "
                        f"portée à {h_req * 2.54:.1f} cm pour éviter les chevauchements")

        fig = _new_figure(Wd, H)
        y = H - top_pad
        if th["banner"]:
            band_h = top_pad + band_pad[0] + rule_h + title_h + subtitle_h + zone_h + band_pad[1]
            fig.add_artist(matplotlib.patches.Rectangle(
                (0, (H - band_h) / H), 1, band_h / H, transform=fig.transFigure,
                facecolor=WSP["black"], edgecolor="none", zorder=0))
            y -= band_pad[0]
            fig.add_artist(matplotlib.patches.Rectangle(
                (margin_l / Wd, (y - 0.045) / H), 0.42 / Wd, 0.045 / H, transform=fig.transFigure,
                facecolor=WSP["red"], edgecolor="none", zorder=1))
            y -= rule_h
            c_title, c_sub, c_zone = WSP["white"], WSP["grey2"], WSP["red"]
        else:
            c_title, c_sub, c_zone = th["text"], th["muted"], th["text"]
        if title_h:
            fig.text(margin_l / Wd, y / H, title, ha="left", va="top", linespacing=1.15,
                     fontsize=fs["title"], fontweight="bold", color=c_title)
            y -= title_h
        if subtitle_h:
            fig.text(margin_l / Wd, y / H, subtitle, ha="left", va="top", linespacing=1.15,
                     fontsize=fs["head"], color=c_sub)
            y -= subtitle_h
        if zone_h:
            fig.text(margin_l / Wd, y / H, zone_name, ha="left", va="top",
                     fontsize=fs["head"], fontweight="bold", color=c_zone)
            y -= zone_h
        if th["banner"]:
            y -= band_pad[1] + after_band
        xmaxes = []
        for ind, m in zip(indicators, metrics):
            ax_top = y - m["head_h"] - m["ann_h"]
            ax_bottom = ax_top - m["bar_h"]
            ax = fig.add_axes([margin_l / Wd, ax_bottom / H, avail / Wd, m["bar_h"] / H])
            if th["head_bar"]:
                hb = fs["head"] * 0.95 / 72
                fig.add_artist(matplotlib.patches.Rectangle(
                    (margin_l / Wd, (ax_top + m["ann_h"] + 0.01) / H), 0.035 / Wd, hb / H,
                    transform=fig.transFigure, facecolor=WSP["red"], edgecolor="none"))
            xmaxes.append(_draw_bullet(fig, ax, ind, s, m, min_xmax))
            leg_top = ax_bottom - m["tick_h"] - m["leg_gap"]
            if ind.contributions and ind.show_legend:
                h, l = _legend_entries(ind, m["ncol"])
                fig.legend(h, l, ncol=m["ncol"], loc="upper left",
                           bbox_to_anchor=(margin_l / Wd, leg_top / H),
                           bbox_transform=fig.transFigure, labelcolor=th["text"],
                           **_legend_kwargs(fs["leg"]))
            y = leg_top - m["leg_h"] - gap
        if footer_h:
            if th["banner"]:
                yl = (y + gap - footer_gap * 0.45) / H
                fig.add_artist(matplotlib.lines.Line2D(
                    [margin_l / Wd, (margin_l + avail) / Wd], [yl, yl], transform=fig.transFigure,
                    color=th["axis"], linewidth=0.6))
            fig.text(margin_l / Wd, (y + gap - footer_gap) / H, footer, ha="left", va="top",
                     fontsize=fs["ann"], color=th["muted"], linespacing=1.2)
        return fig, note, xmaxes


def _for_plot(ind, s):
    """Masque les contributions nulles de la légende si l'option est active (total inchangé)."""
    if s.hide_zero_legend:
        return replace(ind, contributions=[c for c in ind.contributions if c.value > 0])
    return ind


def build_figure(indicators, s, height_cm=0.0, title="", subtitle="", zone_name="", footer="", common_scale=False):
    """Construit une figure avec un bloc par indicateur. Retourne (figure, note sur la hauteur).
    common_scale : même axe pour tous les blocs (comparaison d'un indicateur entre zones)."""
    indicators = [_for_plot(i, s) for i in indicators]
    fig, note, xmaxes = _build(indicators, s, height_cm, title, subtitle, zone_name, footer, 0.0)
    if common_scale and len(xmaxes) > 1 and max(xmaxes) > min(xmaxes) * (1 + 1e-9):
        fig, note, _ = _build(indicators, s, height_cm, title, subtitle, zone_name, footer, max(xmaxes))
    return fig, note


def _join_title(*parts):
    return " — ".join(p.strip() for p in parts if p and p.strip())


def expand_items(k, ind, prefix=""):
    """Blocs à tracer pour un indicateur validé. DH : un bloc par groupe, chacun avec son
    propre DHmax. Autres indicateurs : un seul bloc. prefix : texte avant le nom (ex. zone)."""
    if k == "DH":
        out = []
        for c in ind.contributions:
            tgt = c.target if c.target else ind.target
            out.append(Indicator("DH", ind.unit, tgt, [Contribution(c.label, c.value, c.color, tgt, c.goal)],
                                 True, goal=c.goal, show_legend=False, heading=_join_title(prefix, f"DH — {c.label}"),
                                 sub=f"({ind.unit})" if prefix and ind.unit else ""))
        return out
    return [replace(ind, heading=prefix or k,
                    sub=f"({ind.unit})" if prefix and ind.unit else "")]


def plan_views(zones_valid, s):
    """Liste des graphiques à produire à partir des zones validées.
    zones_valid : [(nom de zone, {indicateur: ValidationResult (indicateurs inclus)})]
    Chaque vue : dict(group, caption, basename, title, subtitle, zone_name, footer, items, height, common).
    Ordre d'une figure : titre, sous-titre, nom de la zone, indicateurs (DH répété par groupe), infos."""
    nz = len(zones_valid)
    prefix = safe_filename(s.file_prefix)
    views = []
    tags = [f"Z{i + 1}_{safe_filename(n)}" for i, (n, _) in enumerate(zones_valid)]
    head = dict(title=s.title, subtitle=s.subtitle, footer=s.footer)

    # 1. Vue d'ensemble : une figure par zone, tous les indicateurs inclus
    for (zname, results), tag in zip(zones_valid, tags):
        if not results or any(not r.ok for r in results.values()):
            continue
        items = [it for k, r in results.items() for it in expand_items(k, r.indicator)]
        views.append(dict(group="zone", caption=f"Vue d'ensemble — {zname}",
                          basename=f"{prefix}_ensemble" + (f"_{tag}" if nz > 1 else ""),
                          zone_name=zname, items=items, height=s.height_overview_cm,
                          common=False, **head))

    # 2. Vue par indicateur : un même indicateur, une barre par zone (à partir de 2 zones)
    if nz > 1:
        for k in INDICATOR_KEYS:
            items = [it for zname, res in zones_valid if k in res and res[k].ok
                     for it in expand_items(k, res[k].indicator, prefix=zname)]
            n_zones = sum(1 for _, res in zones_valid if k in res and res[k].ok)
            if n_zones >= 2:
                views.append(dict(group="indicator", caption=f"{k} — comparaison des zones",
                                  basename=f"{prefix}_{safe_filename(k)}_zones",
                                  zone_name=f"Comparaison des zones — {k}", items=items,
                                  height=s.height_overview_cm, common=s.common_scale, **head))

    # 3. Vues individuelles : une figure par zone et par indicateur (DH : tous ses groupes)
    for (zname, results), tag in zip(zones_valid, tags):
        for k, r in results.items():
            if r.ok:
                views.append(dict(group="single", caption=_join_title(k, zname),
                                  basename=f"{prefix}_" + (f"{tag}_" if nz > 1 else "") + safe_filename(k),
                                  zone_name=zname, items=expand_items(k, r.indicator),
                                  height=s.height_single_cm, common=False, **head))
    return views


# ============================================================
# 4 bis. BILAN BÂTIMENT (anneaux) — un anneau par indicateur bâtiment et par groupe DH
# ============================================================

def fmt_gap(gap, sep):
    """Écart au seuil max en % : « −25 % », « +5 % », « 0 % »."""
    p = round(100 * gap)
    if p == 0:
        return "0\u00a0%"
    return ("+" if p > 0 else "\u2212") + fmt_num(abs(p), 0, sep) + "\u00a0%"


def _ring_cols(avail_in):
    return int(min(5, max(3, avail_in // 1.15)))


def _ring_metrics(s, avail_in):
    fs = s.font_size
    ncol = _ring_cols(avail_in)
    colw = avail_in / ncol
    r = min(colw * 0.34, 0.50 * fs / 9)
    return dict(ncol=ncol, colw=colw, r=r, fs=fs,
                chars=max(8, int(colw * 72 / (fs * 0.56))))


def _ring_cell_h(m, labels, small=False):
    fs = m["fs"]
    lab_fs = fs * (0.85 if small else 1.0)
    nl = max([len(textwrap.wrap(l, m["chars"])) or 1 for l in labels] + [1])
    return nl * lab_fs * 1.25 / 72 + 0.07 + 2 * m["r"] * (0.8 if small else 1.0) + 0.07 + fs * 2.6 / 72


def _draw_ring(ax, cx, ytop, item, m, s, th, mode, small=False, nlines=1):
    """Dessine un anneau dont le haut de la cellule est en ytop (pouces, repère figure)."""
    st = BILAN_STATUS[mode]
    fs = m["fs"]
    acc = BILAN_ACCENT[mode].get(item.key, BILAN_ACCENT[mode]["DH"])
    r = m["r"] * (0.8 if small else 1.0)
    lab_fs = fs * (0.85 if small else 1.0)
    lab = item.label.upper() if th["upper"] else item.label
    lines = textwrap.wrap(lab, m["chars"]) or [""]
    y = ytop
    ax.text(cx, y - (max(nlines, len(lines)) - len(lines)) * lab_fs * 1.25 / 72, "\n".join(lines),
            ha="center", va="top", fontsize=lab_fs, fontweight="bold",
            color=th["text"], linespacing=1.1)
    y -= max(nlines, len(lines)) * lab_fs * 1.25 / 72 + 0.07   # anneaux alignés sur la ligne
    cy = y - r
    ko = not item.ok
    frac = 0.0 if item.max <= 0 else min(max(item.value / item.max, 0.0), 1.0)
    w = r * 0.22
    ax.add_patch(matplotlib.patches.Wedge((cx, cy), r, 0, 360, width=w, facecolor=st["track"],
                                          edgecolor="none"))
    if frac > 0:
        ax.add_patch(matplotlib.patches.Wedge((cx, cy), r, 90 - 360 * frac, 90, width=w,
                                              facecolor=st["ko"] if ko else acc, edgecolor="none"))
    if ko:   # dépassement : arc fin extérieur proportionnel à l'excès
        ex = min(item.gap, 1.0)
        ax.add_patch(matplotlib.patches.Wedge((cx, cy), r + w * 0.55, 90 - 360 * ex, 90, width=w * 0.35,
                                              facecolor=st["ko"], edgecolor="none"))
    txt = fmt_gap(item.gap, s.decimal_sep)
    inner = 2 * (r - w) * 0.92
    p_fs = min(fs * (1.6 if not small else 1.35), inner * 72 / (0.62 * max(len(txt), 4)))
    ax.text(cx, cy + r * 0.07, txt, ha="center", va="center", fontsize=p_fs, fontweight="bold",
            color=st["ko"] if ko else th["text"])
    ax.text(cx, cy - r * 0.36, "vs max", ha="center", va="center", fontsize=fs * 0.66 * (0.9 if small else 1),
            color=th["muted"])
    y = cy - r - 0.07
    ax.text(cx, y, f"{fmt_target(item.value, s.decimals, s.decimal_sep)} / "
                   f"{fmt_target(item.max, s.decimals, s.decimal_sep)}",
            ha="center", va="top", fontsize=fs * 0.95, color=th["text"])
    ax.text(cx, y - fs * 1.35 / 72, item.unit, ha="center", va="top", fontsize=fs * 0.72, color=th["muted"])


def _conformity_bullet(ind, s):
    """Prépare un bullet graph du graphique de conformité : couleur de la barre unique (résultat
    bâtiment saisi directement), écart au seuil max affiché après le titre."""
    mode = "wsp" if s.brand else "base"
    acc = BILAN_ACCENT[mode].get(ind.key, BILAN_ACCENT[mode]["DH"])
    contribs = [replace(c, color=c.color or acc) for c in ind.contributions]
    gap = (ind.total - ind.target) / ind.target if ind.target > 0 else 0.0
    ko = ind.total > ind.target * (1 + 1e-9)
    return _for_plot(replace(ind, contributions=contribs, heading=ind.heading or ind.key,
                             tag=f"{fmt_gap(gap, s.decimal_sep)} vs max", tag_ko=ko), s)


def _bilan_blocks(bilans, s, show_bld_names, style="rings"):
    """Liste ordonnée des blocs à dessiner : (type, données).
    style « rings » : anneaux ; « bullets » : bullet graphs (un bloc par indicateur)."""
    blocks = []
    for b in bilans:
        if show_bld_names:
            blocks.append(("bld", b.name))
        top = b.items if style == "rings" else b.bullets
        dh = b.dh if style == "rings" else b.dh_bullets
        if top:
            blocks.append(("sub", "Indicateurs vérifiés au niveau du bâtiment"))
            if style == "rings":
                blocks.append(("rings", (top, False)))
            else:
                blocks += [("bullet", _conformity_bullet(i, s)) for i in top]
        if dh:
            blocks.append(("sub", "Confort d'été — DH vérifié par groupe"))
            if style == "rings":
                blocks.append(("rings", (dh, True)))
            else:
                blocks += [("bullet", _conformity_bullet(i, s)) for i in dh]
    return blocks


def build_bilan_figure(bilans, s, title="", subtitle="", head_name="", footer="", conform_text=""):
    """Figure « Bilan » : en-tête identique aux autres graphiques, bandeau de conformité
    (si conform_text), anneaux par bâtiment. bilans : [BuildingBilan]."""
    return _build_summary_figure(bilans, s, title, subtitle, head_name, footer, conform_text, "rings")


def build_conformity_figure(bilans, s, title="", subtitle="", head_name="", footer="", conform_text=""):
    """Figure « Conformité RE2020 » : même organisation que le bilan bâtiment (bâtiment, puis
    indicateurs bâtiment, puis DH par groupe), avec les bullet graphs des vues d'ensemble."""
    return _build_summary_figure(bilans, s, title, subtitle, head_name, footer, conform_text, "bullets")


def _build_summary_figure(bilans, s, title, subtitle, head_name, footer, conform_text, style):
    with _rc():
        fs = _font_sizes(s.font_size)
        th = theme(s)
        mode = "wsp" if s.brand else "base"
        st = BILAN_STATUS[mode]
        if th["upper"]:
            title = (title or "").upper()
            head_name = (head_name or "").upper()
        Wd = s.width_cm / 2.54
        margin_l, margin_r = 0.30, 0.40
        avail = Wd - margin_l - margin_r
        rm = _ring_metrics(s, avail)
        top_pad, bottom_pad = 0.12, 0.10
        title = _wrap_title(title, avail, fs["title"])
        subtitle = _wrap_title(subtitle, avail, fs["head"])
        head_name = _wrap_title(head_name, avail, fs["head"])
        footer = _wrap_title(footer, avail, fs["ann"])
        title_h = (fs["title"] * (1.2 * (title.count("\n") + 1) + 0.35) / 72) if title else 0.0
        subtitle_h = (fs["head"] * (1.2 * (subtitle.count("\n") + 1) + 0.25) / 72) if subtitle else 0.0
        zone_h = (fs["head"] * 1.7 / 72) if head_name else 0.0
        footer_gap = 0.25
        footer_h = (footer_gap + fs["ann"] * (1.25 * (footer.count("\n") + 1) + 0.4) / 72) if footer else 0.0
        rule_h = 0.12 if th["banner"] else 0.0
        band_pad = (0.06, 0.12) if th["banner"] else (0.0, 0.0)
        after_band = 0.16 if th["banner"] else 0.06
        ok_h = (s.font_size * 3.6 / 72 + 0.10) if conform_text else 0.0
        ok_gap = 0.18 if conform_text else 0.0
        legend_txt = "Écart = (valeur − seuil max) / seuil max. Négatif = sous le seuil réglementaire."
        if style == "bullets":
            legend_txt = ("Trait plein = seuil max réglementaire" + (", pointillé = cible" if th["goal_dash"]
                                                                     else ", trait gris = cible")
                          + ". " + legend_txt)
        legend_txt = textwrap.fill(legend_txt, max(40, int(avail * 72 / (fs["ann"] * 0.9 * 0.5))))
        legend_h = fs["ann"] * 0.9 * 1.3 * (legend_txt.count("\n") + 1) / 72 + 0.10
        b_gap = max(0.20, s.font_size * 2.2 / 72)      # espace entre deux bullet graphs

        blocks = _bilan_blocks(bilans, s, len(bilans) > 1, style)
        heights = []
        for bi, (kind, data) in enumerate(blocks):
            if kind == "bullet":
                m = _block_metrics(data, s, avail)
                blocks[bi] = (kind, (data, m))
                heights.append(m["total"] + b_gap)
            elif kind == "bld":
                heights.append(fs["head"] * 1.9 / 72 + 0.06)
            elif kind == "sub":
                heights.append(fs["head"] * 0.95 * 1.75 / 72)
            elif kind == "rings":
                items, small = data
                nrow = math.ceil(len(items) / rm["ncol"])
                rows = [items[i * rm["ncol"]:(i + 1) * rm["ncol"]] for i in range(nrow)]
                heights.append(sum(_ring_cell_h(rm, [it.label.upper() if th["upper"] else it.label
                                                     for it in row], small) + 0.12 for row in rows) + 0.06)
            else:
                heights.append(fs["ann"] * 1.6 / 72 * (1 + len(data) // 110) + 0.04)
        H = (rule_h + sum(band_pad) + after_band + top_pad + title_h + subtitle_h + zone_h + ok_h + ok_gap
             + sum(heights) + legend_h + footer_h + bottom_pad)

        fig = _new_figure(Wd, H)
        ax = fig.add_axes([0, 0, 1, 1])
        ax.set_xlim(0, Wd)
        ax.set_ylim(0, H)
        ax.axis("off")
        y = H - top_pad
        if th["banner"]:
            band_h = top_pad + band_pad[0] + rule_h + title_h + subtitle_h + zone_h + band_pad[1]
            ax.add_patch(matplotlib.patches.Rectangle((0, H - band_h), Wd, band_h,
                                                      facecolor=WSP["black"], edgecolor="none"))
            y -= band_pad[0]
            ax.add_patch(matplotlib.patches.Rectangle((margin_l, y - 0.045), 0.42, 0.045,
                                                      facecolor=WSP["red"], edgecolor="none"))
            y -= rule_h
            c_title, c_sub, c_zone = WSP["white"], WSP["grey2"], WSP["red"]
        else:
            c_title, c_sub, c_zone = th["text"], th["muted"], th["text"]
        if title_h:
            ax.text(margin_l, y, title, ha="left", va="top", linespacing=1.15,
                    fontsize=fs["title"], fontweight="bold", color=c_title)
            y -= title_h
        if subtitle_h:
            ax.text(margin_l, y, subtitle, ha="left", va="top", linespacing=1.15,
                    fontsize=fs["head"], color=c_sub)
            y -= subtitle_h
        if zone_h:
            ax.text(margin_l, y, head_name, ha="left", va="top", fontsize=fs["head"],
                    fontweight="bold", color=c_zone)
            y -= zone_h
        if th["banner"]:
            y -= band_pad[1]
        y -= after_band

        # Bandeau de conformité (uniquement si tous les indicateurs sont sous le seuil)
        if conform_text:
            ax.add_patch(matplotlib.patches.Rectangle((margin_l, y - ok_h), avail, ok_h,
                                                      facecolor=st["ok_bg"], edgecolor="none"))
            if th["banner"]:
                ax.add_patch(matplotlib.patches.Rectangle((margin_l, y - ok_h), 0.07, ok_h,
                                                          facecolor=WSP["red"], edgecolor="none"))
            cyc = y - ok_h / 2
            rr = ok_h * 0.30
            xc = margin_l + 0.18 + rr
            ax.add_patch(matplotlib.patches.Circle((xc, cyc), rr, facecolor="none",
                                                   edgecolor=WSP["red"] if th["banner"] else st["ok_fg"],
                                                   linewidth=1.6))
            ax.plot([xc - rr * 0.45, xc - rr * 0.08, xc + rr * 0.50],
                    [cyc + rr * 0.02, cyc - rr * 0.36, cyc + rr * 0.40],
                    color=WSP["red"] if th["banner"] else st["ok_fg"], linewidth=2.0,
                    solid_capstyle="round", solid_joinstyle="round")
            txt = conform_text.upper() if th["upper"] else conform_text
            t_fs = s.font_size * 1.55
            t = ax.text(xc + rr + 0.14, cyc, txt, ha="left", va="center", fontsize=t_fs,
                        fontweight="bold", color=st["ok_fg"])
            r_ = fig.canvas.get_renderer()
            wmax = (margin_l + avail - 0.1 - (xc + rr + 0.14)) * fig.dpi
            bw = t.get_window_extent(r_).width
            if bw > wmax:
                t.set_fontsize(t_fs * wmax / bw)
            y -= ok_h + ok_gap

        for (kind, data), h in zip(blocks, heights):
            if kind == "bld":
                nm = data.upper() if th["upper"] else data
                if th["head_bar"]:
                    ax.add_patch(matplotlib.patches.Rectangle((margin_l, y - fs["head"] * 1.05 / 72), 0.035,
                                                              fs["head"] * 0.95 / 72, facecolor=WSP["red"],
                                                              edgecolor="none"))
                ax.text(margin_l + (0.08 if th["head_bar"] else 0), y, nm, ha="left", va="top",
                        fontsize=fs["head"] * 1.05, fontweight="bold", color=th["text"])
                yl = y - fs["head"] * 1.5 / 72
                ax.plot([margin_l, margin_l + avail], [yl, yl], color=th["axis"], linewidth=0.6)
            elif kind == "sub":
                t = data.upper() if th["upper"] else data
                ax.text(margin_l, y, t, ha="left", va="top", fontsize=fs["head"] * 0.85,
                        fontweight="bold", color=th["muted"])
            elif kind == "bullet":
                ind, m = data
                ax_top = y - m["head_h"] - m["ann_h"]
                ax_bottom = ax_top - m["bar_h"]
                bax = fig.add_axes([margin_l / Wd, ax_bottom / H, avail / Wd, m["bar_h"] / H])
                if th["head_bar"]:
                    ax.add_patch(matplotlib.patches.Rectangle((margin_l, ax_top + m["ann_h"] + 0.01), 0.035,
                                                              fs["head"] * 0.95 / 72, facecolor=WSP["red"],
                                                              edgecolor="none"))
                _draw_bullet(fig, bax, ind, s, m)
                leg_top = ax_bottom - m["tick_h"] - m["leg_gap"]
                if ind.contributions and ind.show_legend:
                    hh, ll = _legend_entries(ind, m["ncol"])
                    fig.legend(hh, ll, ncol=m["ncol"], loc="upper left",
                               bbox_to_anchor=(margin_l / Wd, leg_top / H),
                               bbox_transform=fig.transFigure, labelcolor=th["text"],
                               **_legend_kwargs(fs["leg"]))
            elif kind == "rings":
                items, small = data
                yy = y
                for i in range(0, len(items), rm["ncol"]):
                    row = items[i:i + rm["ncol"]]
                    labs = [it.label.upper() if th["upper"] else it.label for it in row]
                    nl = max(len(textwrap.wrap(l, rm["chars"])) or 1 for l in labs)
                    for j, it in enumerate(row):
                        cx = margin_l + rm["colw"] * (j + 0.5)
                        _draw_ring(ax, cx, yy, it, rm, s, th, mode, small, nl)
                    yy -= _ring_cell_h(rm, labs, small) + 0.12
            else:
                ax.text(margin_l, y, textwrap.fill(data, 110), ha="left", va="top", fontsize=fs["ann"],
                        color=BILAN_STATUS[mode]["ko"], style="italic")
            y -= h
        ax.text(margin_l, y, legend_txt, ha="left", va="top", fontsize=fs["ann"] * 0.9,
                color=th["muted"], style="italic")
        y -= legend_h
        if footer_h:
            if th["banner"]:
                yl = y - footer_gap * 0.45
                ax.plot([margin_l, margin_l + avail], [yl, yl], color=th["axis"], linewidth=0.6)
            ax.text(margin_l, y - footer_gap, footer, ha="left", va="top", fontsize=fs["ann"],
                    color=th["muted"], linespacing=1.2)
        return fig, None


def plan_bilan_views(bilans, s, project_ok):
    """Vues « Bilan » : une vue projet (tous les bâtiments) + une vue par bâtiment s'il y en a
    plusieurs. Le bandeau de conformité n'apparaît que si tout est sous le seuil."""
    prefix = safe_filename(s.file_prefix)
    head = dict(title=s.title, subtitle=s.subtitle, footer=s.footer)
    views = []
    if not bilans:
        return views
    if len(bilans) == 1:
        b = bilans[0]
        views.append(dict(group="bilan", kind="bilan", caption=f"Bilan — {b.name}",
                          basename=f"{prefix}_bilan", zone_name=f"{b.name} — Bilan bâtiment",
                          bilans=[b], conform="Le projet est conforme à la RE2020" if b.conform else "",
                          scope="project", **head))
        return views
    views.append(dict(group="bilan", kind="bilan", caption="Bilan — projet (tous les bâtiments)",
                      basename=f"{prefix}_bilan_projet", zone_name="Bilan du projet",
                      bilans=bilans, conform="Le projet est conforme à la RE2020" if project_ok else "",
                      scope="project", **head))
    for i, b in enumerate(bilans):
        views.append(dict(group="bilan", kind="bilan", caption=f"Bilan — {b.name}",
                          basename=f"{prefix}_bilan_B{i + 1}_{safe_filename(b.name)}",
                          zone_name=f"{b.name} — Bilan bâtiment", bilans=[b],
                          conform=f"{b.name} : bâtiment conforme à la RE2020" if b.conform else "",
                          scope="building", **head))
    return views


def plan_conformity_views(bilans, s, project_ok):
    """Vues « Conformité RE2020 » (bullet graphs organisés comme le bilan) : même découpage
    que le bilan (projet + un graphique par bâtiment s'il y en a plusieurs)."""
    views = []
    for v in plan_bilan_views(bilans, s, project_ok):
        v = dict(v, group="conformite", kind="conformite",
                 caption=v["caption"].replace("Bilan", "Conformité RE2020", 1),
                 basename=v["basename"].replace("_bilan", "_conformite", 1),
                 zone_name=v["zone_name"].replace("Bilan bâtiment", "Conformité RE2020")
                                         .replace("Bilan du projet", "Conformité RE2020 du projet"))
        views.append(v)
    return views


def build_view(view, s):
    if view.get("kind") == "conformite":
        return build_conformity_figure(view["bilans"], s, view["title"], view.get("subtitle", ""),
                                       view.get("zone_name", ""), view.get("footer", ""), view.get("conform", ""))
    if view.get("kind") == "bilan":
        return build_bilan_figure(view["bilans"], s, view["title"], view.get("subtitle", ""),
                                  view.get("zone_name", ""), view.get("footer", ""), view.get("conform", ""))
    return build_figure(view["items"], s, view["height"], view["title"], view.get("subtitle", ""), view.get("zone_name", ""), view.get("footer", ""), view["common"])


# ============================================================
# 4 ter. FAMILLES DE GRAPHIQUES (choix de l'utilisateur, calcul à la demande)
# ============================================================
# Synthèse : bilan + conformité du projet (ou du bâtiment unique)
# Bâtiment(s) : bilan + conformité de chaque bâtiment (projet à plusieurs bâtiments)
# Zone(s) : vue d'ensemble par zone ; Indicateurs : un indicateur comparé entre zones ;
# Individuelles : une figure par zone et par indicateur.
FAMILIES = [("synth", "Synthèse", "Bilan et conformité RE2020 du projet"),
            ("bld", "Bâtiment(s)", "Bilan et conformité de chaque bâtiment (projet à plusieurs bâtiments)"),
            ("zone", "Zone(s)", "Vue d'ensemble de chaque zone (tous les indicateurs)"),
            ("ind", "Indicateurs", "Un indicateur comparé entre les zones (au moins 2 zones)"),
            ("single", "Individuelles", "Une figure par zone et par indicateur")]
FAMILY_KEYS = [k for k, _, _ in FAMILIES]
FAMILY_LABEL = {k: lab for k, lab, _ in FAMILIES}
DEFAULT_FAMILIES = {"synth", "bld"}


def view_family(v):
    """Famille d'une vue planifiée."""
    g = v.get("group")
    if g in ("bilan", "conformite"):
        return "synth" if v.get("scope", "project") == "project" else "bld"
    return {"zone": "zone", "indicator": "ind", "single": "single"}.get(g, "single")


def plan_family_views(s, zones_valid, bilans, project_ok, families):
    """Planifie les vues des familles demandées, dans l'ordre des familles.
    Aucune figure n'est construite ici (planification seule, très rapide)."""
    families = set(families)
    views = []
    if families & {"synth", "bld"}:
        bv = plan_bilan_views(bilans, s, project_ok)
        cv = plan_conformity_views(bilans, s, project_ok)
        for b, c in zip(bv, cv):          # bilan puis conformité, pour chaque portée
            views += [b, c]
    if families & {"zone", "ind", "single"}:
        views += plan_views(zones_valid, s)
    order = {k: i for i, k in enumerate(FAMILY_KEYS)}
    views = [v for v in views if view_family(v) in families]
    views.sort(key=lambda v: order[view_family(v)])   # tri stable
    return views
```

```python
# ============================================================
# 5. EXPORT DES FIGURES ET SAUVEGARDE JSON
# ============================================================

def _bbox_kwargs(fig):
    """Recadrage 'tight' uniquement si un élément dépasse du cadre (sécurité anti-coupure)."""
    r = fig.canvas.get_renderer()
    tb = fig.get_tightbbox(r)
    w, h = fig.get_size_inches()
    tol = 0.01
    if tb.x0 < -tol or tb.y0 < -tol or tb.x1 > w + tol or tb.y1 > h + tol:
        return {"bbox_inches": "tight", "pad_inches": 0.08}
    return {}


def figure_bytes(fig, fmt="png", dpi=300, transparent=False):
    buf = io.BytesIO()
    with _rc():
        kw = _bbox_kwargs(fig)
        if transparent:
            fig.savefig(buf, format=fmt, dpi=dpi, transparent=True, **kw)
        else:
            fig.savefig(buf, format=fmt, dpi=dpi, facecolor="white", **kw)
    return buf.getvalue()


def safe_filename(text):
    t = unicodedata.normalize("NFKD", str(text)).encode("ascii", "ignore").decode()
    t = re.sub(r"[^A-Za-z0-9_-]+", "_", t).strip("_")
    return t or "graphique"


def export_figure(fig, basename, s):
    """Écrit la figure dans tous les formats sélectionnés. Retourne la liste des chemins."""
    os.makedirs(s.export_dir, exist_ok=True)
    paths = []
    for fmt in s.formats:
        path = os.path.join(s.export_dir, f"{basename}.{fmt}")
        data = figure_bytes(fig, fmt=fmt, dpi=s.dpi, transparent=s.transparent)
        with open(path, "wb") as f:
            f.write(data)
        paths.append(os.path.abspath(path))
    return paths


def project_json_bytes(state):
    """Projet sérialisé en JSON (UTF-8), sans écriture disque."""
    out = json.loads(json.dumps(state))   # copie profonde
    for zone in out["zones"]:
        for raw in zone["indicators"].values():
            for fld in ("target", "goal"):
                v, err = parse_number(raw.get(fld, ""))
                raw[fld] = v if err is None else raw.get(fld, "")
            for c in raw["contributions"]:
                v, err = parse_number(c.get("value", ""))
                c["value"] = v if err is None else c.get("value", "")
    out["app_version"] = APP_VERSION
    out["saved_at"] = now_local().isoformat(timespec="seconds")
    return json.dumps(out, ensure_ascii=False, indent=2).encode("utf-8")


def save_project_json(state, path):
    path = path.strip()
    if not path:
        raise ValueError("indiquez un nom de fichier JSON.")
    if not path.lower().endswith(".json"):
        path += ".json"
    folder = os.path.dirname(path)
    if folder:
        os.makedirs(folder, exist_ok=True)
    with open(path, "wb") as f:
        f.write(project_json_bytes(state))
    return os.path.abspath(path)


def project_from_bytes(data):
    """Relit un projet JSON téléversé."""
    try:
        return normalize_state(json.loads(bytes(data).decode("utf-8-sig")))
    except (UnicodeDecodeError, json.JSONDecodeError) as e:
        raise ValueError(f"JSON illisible : {e}")


def figures_zip_bytes(files):
    """files : [(nom de fichier, octets)] -> archive ZIP en mémoire."""
    buf = io.BytesIO()
    with zipfile.ZipFile(buf, "w", zipfile.ZIP_DEFLATED) as z:
        for name, data in files:
            z.writestr(name, data)
    return buf.getvalue()


def load_project_json(path):
    path = path.strip()
    if not path:
        raise ValueError("indiquez le chemin du fichier JSON.")
    if not os.path.isfile(path):
        raise FileNotFoundError(f"fichier introuvable : {os.path.abspath(path)}")
    with open(path, "r", encoding="utf-8") as f:
        try:
            data = json.load(f)
        except json.JSONDecodeError as e:
            raise ValueError(f"JSON illisible (ligne {e.lineno}) : {e.msg}")
    return normalize_state(data), os.path.abspath(path)
```

```python
# ============================================================
# 5 bis. ÉCHANGE AVEC EXCEL PAR COPIER-COLLER (modèle « clé | valeur »)
# ============================================================
# Collage : une clé par ligne (ex. « BbioChaud_widget »), puis une valeur par zone :
#   colonne B = 1re zone, colonne C = 2e zone, etc.
# Une ligne « zone_widget | Zone A | Zone B » (facultative) associe les colonnes aux zones par nom.
# Seules les clés collées sont mises à jour : on peut coller tout le tableau ou quelques lignes.

EXCEL_TITLE_KEY = "titre"
EXCEL_SUBTITLE_KEY = "soustitre"
EXCEL_FOOTER_KEY = "infos"
EXCEL_ZONE_KEY = "zone"
EXCEL_CEPNR_EQUALS_CEP_KEY = "cepnr_equals_cep"
EXCEL_BUILDING_KEY = "batiment"     # nom du bâtiment de la zone (une valeur par colonne de zone)
EXCEL_SURFACE_KEY = "surface"       # surface de la zone (pondération du bilan bâtiment automatique)
# Résultats saisis directement au niveau bâtiment (facultatif) : (indicateur, champ)
BAT_RESULT_KEYS = {
    "bat_bbio": ("Bbio", "value"), "bat_bbiomax": ("Bbio", "max"),
    "bat_cep": ("CEP", "value"), "bat_cepmax": ("CEP", "max"),
    "bat_cepnr": ("CEPnr", "value"), "bat_cepnrmax": ("CEPnr", "max"),
    "bat_ice": ("ICénergie", "value"), "bat_icemax": ("ICénergie", "max"),
    "bat_icc": ("ICconstruction", "value"), "bat_iccmax": ("ICconstruction", "max"),
}
# Clés du modèle reconnues mais sans équivalent dans ce notebook (signalées, ignorées)
EXCEL_IGNORED_KEYS = {"nbrange", "heightbar", "heightfond", "graduationx", "alphafond"}
# Groupes DH : DH_Gi_widget (valeur), DH_Gi_nom_widget (libellé), DH_Gi_max_widget (DHmax du groupe)
DH_GROUP_EXTRA_KEYS = {f"dh_g{i}_{s}" for i in range(1, 9) for s in ("nom", "max", "cible")}
# Lignes d'en-tête / de section du modèle Excel, ignorées silencieusement
EXCEL_HEADER_KEYS = {"clé", "cle", "clé(nepasmodifier)"}


def _excel_key(v):
    """Normalise une clé : minuscules, sans espaces ni suffixe _widget ; IcC_lot1 -> icc_lot01."""
    if not isinstance(v, str):
        return None
    k = v.strip().lower().replace(" ", "")
    if k.endswith("_widget"):
        k = k[:-7]
    m = re.fullmatch(r"icc_lot0*(\d+)", k)
    if m:
        k = f"icc_lot{int(m.group(1)):02d}"
    return k or None


def _known_keys():
    keys = ({EXCEL_TITLE_KEY, EXCEL_SUBTITLE_KEY, EXCEL_FOOTER_KEY, EXCEL_ZONE_KEY, EXCEL_CEPNR_EQUALS_CEP_KEY,
             EXCEL_BUILDING_KEY, EXCEL_SURFACE_KEY} | set(BAT_RESULT_KEYS) | EXCEL_IGNORED_KEYS
            | DH_GROUP_EXTRA_KEYS)
    for spec in EXCEL_MAP.values():
        keys |= {spec["show"], spec["max"], spec["goal"]} | {c[0] for c in spec["contribs"]}
    return keys


def _excel_bool(v, default=True):
    s = str(v).strip().lower() if v is not None else ""
    if not s:
        return default
    if s in ("1", "oui", "o", "vrai", "true", "yes", "x"):
        return True
    if s in ("0", "non", "n", "faux", "false", "no"):
        return False
    num, err = parse_number(s)
    return (num != 0) if err is None else default


def parse_pasted_table(text):
    """Découpe un tableau collé depuis Excel. Retourne (lignes reconnues, lignes inconnues).
    Ligne reconnue : (clé normalisée, clé d'origine, [valeurs par colonne])."""
    known = _known_keys()
    rows, unknown = [], []
    for line in str(text or "").splitlines():
        if not line.strip():
            continue
        if "\t" in line:                      # copie Excel standard
            cells = line.split("\t")
        elif ";" in line:                     # CSV « français »
            cells = line.split(";")
        else:                                 # saisie manuelle « clé valeur »
            cells = line.strip().split(None, 1)
        cells = [c.strip().strip('"').strip() for c in cells]
        j = next((i for i, c in enumerate(cells) if c), None)
        if j is None:
            continue
        if cells[j].startswith(("#", "▸")):     # ligne de section du modèle
            continue
        k = _excel_key(cells[j])
        if k in EXCEL_HEADER_KEYS:
            continue
        vals = cells[j + 1:]
        while vals and vals[-1] == "":
            vals.pop()
        if k in known:
            rows.append((k, cells[j], vals))
        else:
            unknown.append(cells[j][:40])
    return rows, unknown


def _merge_contributions(key, raw, pairs, warnings, zname):
    """Met à jour les contributions d'un indicateur avec les clés collées.
    Hors DH : structure fixe du modèle (toutes les contributions, une contribution nulle n'est
    pas tracée) ; seules les valeurs collées changent, libellés et couleurs sont conservés.
    DH : un groupe par clé DH_Gi collée ou déjà saisie (DH = 0 est un résultat)."""
    spec = EXCEL_MAP[key]
    existing = raw["contributions"][:raw["n_active"]]
    if key != "DH":
        new = canonical_contribs(key, existing)
        for c in new:
            if c["key"] in pairs:
                c["value"] = str(pairs[c["key"]])
                _, err = parse_number(c["value"])
                if err:
                    warnings.append(f"{zname} / {key} — {excel_name(c['key'])} : {err} (à corriger dans l'onglet).")
        raw["contributions"] = new
        raw["n_active"] = max_contrib(key)
        return
    by_key = {c["key"]: c for c in existing if c.get("key")}
    new = []
    for ck, label in spec["contribs"]:
        if ck in pairs:
            value = pairs[ck]
            _, err = parse_number(value)
            if err:
                warnings.append(f"{zname} / {key} — {excel_name(ck)} : {err} (à corriger dans l'onglet).")
        elif ck in by_key:
            value = by_key[ck]["value"]
        else:
            continue
        base = by_key.get(ck, {})
        new.append({"key": ck, "label": base.get("label") or label, "value": str(value),
                    "max": base.get("max", ""), "goal": base.get("goal", ""),
                    "color": base.get("color") or _spec_color(key, ck)})
    if not new:   # au moins un groupe (valeur nulle)
        ck, label = spec["contribs"][0]
        new = [{"key": ck, "label": label, "value": "0", "max": "", "goal": "", "color": _spec_color(key, ck)}]
    raw["contributions"] = _pad(new, key)
    raw["n_active"] = min(len(new), max_contrib(key))


def apply_pasted_table(state, text, start_zone=0):
    """Applique un tableau collé à l'état. start_zone : zone de la 1re colonne de valeurs
    (-1 = nouvelle zone). Retourne (nouvel état, compte rendu)."""
    rows, unknown = parse_pasted_table(text)
    if not rows:
        raise ValueError("aucune clé du modèle reconnue dans le texte collé "
                         "(ex. « Bbiomax_widget<TAB>180,7 »).")
    state = normalize_state(copy.deepcopy(state))
    zones = state["zones"]
    rep = {"updated": [], "created": [], "warnings": [], "ignored": [], "unknown": unknown}
    buildings = state["buildings"]

    table = {}
    for k, orig, vals in rows:
        if k in table:
            rep["warnings"].append(f"Clé « {orig} » collée plusieurs fois : dernière ligne retenue.")
        table[k] = (orig, vals)
    rep["ignored"] = [table[k][0] for k in table if k in EXCEL_IGNORED_KEYS]

    if EXCEL_TITLE_KEY in table:
        t = next((v for v in table[EXCEL_TITLE_KEY][1] if v), "")
        if t:
            state["settings"]["title"] = t
            rep["updated"].append("Titre du graphique")
    if EXCEL_SUBTITLE_KEY in table:
        t = next((v for v in table[EXCEL_SUBTITLE_KEY][1] if v), "")
        state["settings"]["subtitle"] = t
        rep["updated"].append("Sous-titre")
    if EXCEL_FOOTER_KEY in table:
        t = next((v for v in table[EXCEL_FOOTER_KEY][1] if v), "")
        state["settings"]["footer"] = t
        rep["updated"].append("Informations de bas de graphique")

    data_keys = [k for k in table if k not in (EXCEL_TITLE_KEY, EXCEL_SUBTITLE_KEY, EXCEL_FOOTER_KEY, EXCEL_ZONE_KEY)
                 and k not in EXCEL_IGNORED_KEYS]
    ncols = max([len(table[k][1]) for k in data_keys] + [0])
    if ncols == 0 and not rep["updated"]:
        raise ValueError("clés reconnues mais aucune valeur : vérifiez que la colonne des "
                         "valeurs a bien été copiée.")

    names = table.get(EXCEL_ZONE_KEY, ("", []))[1]
    n_before = len(zones)
    touched_by_zone = {}
    base = len(zones) if start_zone is None or start_zone < 0 else min(start_zone, len(zones))
    for j in range(ncols):
        name = names[j].strip() if j < len(names) else ""
        bcol = table.get(EXCEL_BUILDING_KEY, ("", []))[1]
        bname_j = bcol[j].strip() if j < len(bcol) else ""
        bi_j = next((i for i, b in enumerate(buildings) if b["name"].casefold() == bname_j.casefold()), None) \
            if bname_j else None
        if name:
            # même nom de zone possible dans plusieurs bâtiments : on cherche d'abord (bâtiment, zone)
            cands = [i for i, z in enumerate(zones) if z["name"].casefold() == name.casefold()]
            if bname_j:
                cands = [i for i in cands if bi_j is not None and int(zones[i].get("building", 0)) == bi_j]
            zi = cands[0] if cands else None
            if zi is None:
                zones.append(new_zone(name, bi_j or 0))
                zi = len(zones) - 1
                rep["created"].append(f"{bname_j} / {name}" if bname_j else name)
        else:
            zi = base + j
            while zi >= len(zones):
                zones.append(new_zone(f"Zone {len(zones) + 1}"))
                rep["created"].append(zones[-1]["name"])
        zone = zones[zi]
        zname = zone["name"]
        pairs = {k: table[k][1][j] for k in data_keys if j < len(table[k][1]) and table[k][1][j] != ""}
        if not pairs:
            continue
        touched = []
        bname = str(pairs.get(EXCEL_BUILDING_KEY, "") or "").strip()
        if bname:
            bi = next((i for i, b in enumerate(buildings) if b["name"].casefold() == bname.casefold()), None)
            if bi is None:
                buildings.append(new_building(bname))
                bi = len(buildings) - 1
                rep["created"].append(f"bâtiment {bname}")
            zone["building"] = bi
            touched.append(f"bâtiment = {buildings[bi]['name']}")
        if EXCEL_SURFACE_KEY in pairs:
            zone["surface"] = pairs[EXCEL_SURFACE_KEY]
            v, err = parse_number(zone["surface"])
            if err or v <= 0:
                rep["warnings"].append(f"{zname} — surface_widget : surface invalide.")
            touched.append("surface")
        bat_pairs = {k: pairs[k] for k in BAT_RESULT_KEYS if k in pairs}
        if bat_pairs:
            b = buildings[zone.get("building", 0)]
            for k, v in bat_pairs.items():
                ind, fld = BAT_RESULT_KEYS[k]
                b["results"][ind][fld] = v
            b["mode"] = "manual"
            touched.append(f"résultats bâtiment ({b['name']})")
        for key, spec in EXCEL_MAP.items():
            raw = zone["indicators"][key]
            changed = False
            if any(ck in pairs for ck, _ in spec["contribs"]):
                _merge_contributions(key, raw, pairs, rep["warnings"], zname)
                changed = True
            if spec["max"] in pairs:
                raw["target"] = pairs[spec["max"]]
                v, err = parse_number(raw["target"])
                if err or v <= 0:
                    rep["warnings"].append(f"{zname} / {key} — {excel_name(spec['max'])} : "
                                           f"valeur maximale invalide (à corriger dans l'onglet).")
                changed = True
            if spec["goal"] in pairs:
                raw["goal"] = pairs[spec["goal"]]
                v, err = parse_number(raw["goal"])
                if err or v <= 0:
                    rep["warnings"].append(f"{zname} / {key} — {excel_name(spec['goal'])} : "
                                           f"cible invalide (à corriger dans l'onglet).")
                changed = True
            if spec["show"] in pairs:
                raw["include"] = _excel_bool(pairs[spec["show"]], True)
                changed = True
            if key == "DH":
                by_key = {c["key"]: c for c in raw["contributions"][:raw["n_active"]] if c.get("key")}
                for gi in range(1, 9):
                    c = by_key.get(f"dh_g{gi}")
                    nk, mk, gk = f"dh_g{gi}_nom", f"dh_g{gi}_max", f"dh_g{gi}_cible"
                    if c is None:
                        if f"dh_g{gi}" not in pairs and (nk in pairs or mk in pairs or gk in pairs):
                            pass   # nom / max sans valeur DH : ignorés
                        continue
                    if nk in pairs:
                        c["label"] = pairs[nk].strip() or f"Groupe {gi}"
                        changed = True
                    if mk in pairs:
                        c["max"] = pairs[mk]
                        v, err = parse_number(c["max"])
                        if err or v <= 0:
                            rep["warnings"].append(f"{zname} / DH — DH_G{gi}_max_widget : "
                                                   f"valeur maximale invalide (à corriger dans l'onglet).")
                        changed = True
                    if gk in pairs:
                        c["goal"] = pairs[gk]
                        changed = True
            if changed:
                touched.append(key + ("" if raw["include"] else " (masqué)"))
        if _excel_bool(pairs.get(EXCEL_CEPNR_EQUALS_CEP_KEY), False):
            cep, cepnr = zone["indicators"]["CEP"], zone["indicators"]["CEPnr"]
            contribs = copy.deepcopy(cep["contributions"])
            for c in contribs:
                if c["key"].startswith("cep") and not c["key"].startswith("cepnr"):
                    c["key"] = "cepnr" + c["key"][3:]
            cepnr["contributions"], cepnr["n_active"] = contribs, cep["n_active"]
            rep["warnings"].append(f"{zname} : cepnr_equals_cep = 1, contributions du CEPnr recopiées "
                                   f"depuis le CEP (cible CEPnr inchangée).")
            if not any(t.startswith("CEPnr") for t in touched):
                touched.append("CEPnr")
        touched_by_zone.setdefault(zi, set()).update(t.split(" ")[0] for t in touched)
        if touched:
            rep["updated"].append(f"{zname} : " + ", ".join(touched))
    # Zones créées par le collage : seuls les indicateurs collés sont inclus
    for zi in range(n_before, len(zones)):
        for key in INDICATOR_KEYS:
            if key not in touched_by_zone.get(zi, set()):
                zones[zi]["indicators"][key]["include"] = False
    removed = prune_empty_buildings(state)
    if removed:
        rep["warnings"].append("Bâtiment(s) vide(s) retiré(s) : " + ", ".join(removed) + ".")
    return state, rep


def _fmt_excel_value(v, sep):
    num, err = parse_number(v)
    if err:
        return str(v).strip()
    s = f"{num:.10g}"
    return s.replace(".", ",") if sep == "," else s


SECTION_TITLES = {"Bbio": "Bbio", "CEP": "CEP", "CEPnr": "CEPnr",
                  "DH": "DH — un bloc par groupe (valeur, nom, DHmax et cible du groupe)",
                  "ICénergie": "ICénergie", "ICconstruction": "ICconstruction"}


def state_to_excel_rows(state, sep=",", sections=False):
    """Lignes « clé | zone 1 | zone 2 … » des valeurs actuelles.
    sections=True : ajoute des lignes de section (« ▸ … », ignorées à l'import).
    Retourne (lignes, avertissements)."""
    zones = state["zones"]
    warnings = []
    blds = state.get("buildings") or [new_building("Bâtiment 1")]
    rows = [["titre_widget", state["settings"].get("title", "")],
            ["sousTitre_widget", state["settings"].get("subtitle", "")],
            ["infos_widget", state["settings"].get("footer", "")],
            ["zone_widget"] + [z["name"] for z in zones],
            ["batiment_widget"] + [blds[min(int(z.get("building", 0)), len(blds) - 1)]["name"] for z in zones],
            ["surface_widget"] + [_fmt_excel_value(z.get("surface", ""), sep) for z in zones]]
    for key, spec in EXCEL_MAP.items():
        if sections:
            rows.append([f"▸ {SECTION_TITLES[key]}"])
        per_zone = []
        for z in zones:
            raw = z["indicators"][key]
            active = raw["contributions"][:raw["n_active"]]
            per_zone.append({c["key"]: c for c in active if c.get("key")})
            manual = [c["label"] or "sans libellé" for c in active if not c.get("key")]
            if manual:
                warnings.append(f"{z['name']} / {key} : contribution(s) sans clé Excel non exportée(s) : "
                                + ", ".join(manual) + ".")
        if key == "DH":
            for gi in range(1, 9):
                ck = f"dh_g{gi}"
                cs = [pz.get(ck) for pz in per_zone]
                rows.append([f"DH_G{gi}_widget"] + [_fmt_excel_value(c["value"], sep) if c else "" for c in cs])
                rows.append([f"DH_G{gi}_nom_widget"] + [c["label"] if c else "" for c in cs])
                rows.append([f"DH_G{gi}_max_widget"] + [_fmt_excel_value(c.get("max", ""), sep) if c else ""
                                                        for c in cs])
                rows.append([f"DH_G{gi}_cible_widget"] + [_fmt_excel_value(c.get("goal", ""), sep) if c else ""
                                                          for c in cs])
        else:
            for ck, _ in spec["contribs"]:
                rows.append([excel_name(ck)] + [_fmt_excel_value(pz[ck]["value"], sep) if ck in pz else "0"
                                                for pz in per_zone])
        rows.append([excel_name(spec["max"])]
                    + [_fmt_excel_value(z["indicators"][key]["target"], sep) for z in zones])
        rows.append([excel_name(spec["goal"])]
                    + [_fmt_excel_value(z["indicators"][key].get("goal", ""), sep) for z in zones])
        rows.append([excel_name(spec["show"])]
                    + ["1" if z["indicators"][key]["include"] else "0" for z in zones])
    rows.append(["cepnr_equals_cep_widget"] + ["0"] * len(zones))
    # Résultats saisis au niveau bâtiment : dans la 1re colonne de zone de chaque bâtiment
    if sections:
        rows.append(["▸ Résultats bâtiment (facultatif) — à remplir dans UNE colonne par bâtiment"])
    first_col = {}
    for j, z in enumerate(zones):
        first_col.setdefault(int(z.get("building", 0)), j)
    for k, (ind, fld) in BAT_RESULT_KEYS.items():
        row = [""] * len(zones)
        for bi, j in first_col.items():
            b = blds[min(bi, len(blds) - 1)]
            if b.get("mode") == "manual":
                row[j] = _fmt_excel_value(b["results"][ind][fld], sep)
        rows.append([excel_name(k)] + row)
    return rows, warnings


def state_to_excel_text(state, sep=","):
    """Tableau texte (tabulations) à coller en A1 dans Excel. Retourne (texte, avertissements)."""
    rows, warnings = state_to_excel_rows(state, sep)
    return "\n".join("\t".join(r) for r in rows), warnings


# ---------- Fichier Excel XLSX : génération et lecture en mémoire

KEY_HELP = {
    "titre_widget": "Titre du graphique (colonne B uniquement)",
    "sousTitre_widget": "Sous-titre (colonne B uniquement)",
    "infos_widget": "Informations en bas du graphique (colonne B uniquement)",
    "zone_widget": "Nom de chaque zone (une colonne par zone)",
    "batiment_widget": "Nom du bâtiment auquel appartient la zone (bâtiment créé s'il n'existe pas)",
    "surface_widget": "Surface de la zone (m²), utilisée pour pondérer le bilan bâtiment si plusieurs zones",
    "bat_Bbio_widget": "Facultatif : résultat Bbio du bâtiment (Pléiades). Remplir une seule colonne par bâtiment",
    "DHmax_widget": "DHmax par défaut, utilisé si un groupe n'a pas de DHmax propre",
    "DHcible_widget": "Cible DH par défaut, utilisée si un groupe n'a pas de cible propre",
    "Bbiocible_widget": "Cible facultative (repère discret en plus du max). Vide = pas de cible",
    "cepnr_equals_cep_widget": "1 = contributions du CEPnr recopiées depuis le CEP",
}


def _to_cell(v):
    """Nombre réel dans Excel quand c'est possible (sinon texte)."""
    if isinstance(v, str):
        n, err = parse_number(v)
        if err is None and v.strip() != "":
            return int(n) if n == int(n) else n
    return v


def excel_template_bytes(state=None):
    """Modèle XLSX pré-rempli avec l'état fourni (une colonne par zone)."""
    state = normalize_state(state or demo_state())
    rows, _ = state_to_excel_rows(state, ".", sections=True)
    nz = len(state["zones"])
    wb = Workbook()
    ws = wb.active
    ws.title = "Saisie RE2020"
    F = "Arial"
    dark, sect, keyfill, inp, white = "1F4E78", "DCE6F1", "F2F2F2", "FFF2CC", "FFFFFF"
    thin = Side(style="thin", color="BFBFBF")
    border = Border(left=thin, right=thin, top=thin, bottom=thin)
    ws.append(["Clé (ne pas modifier)"] + [f"Zone {i + 1}" for i in range(nz)])
    for c in ws[1]:
        c.font = Font(name=F, bold=True, color=white)
        c.fill = PatternFill("solid", fgColor=dark)
        c.alignment = Alignment(horizontal="center", vertical="center")
    for r in rows:
        if r[0].startswith("▸"):
            ws.append([r[0]])
            row = ws.max_row
            for col in range(1, nz + 2):
                cell = ws.cell(row, col)
                cell.fill = PatternFill("solid", fgColor=sect)
                cell.font = Font(name=F, bold=True, color=dark)
            continue
        vals = [_to_cell(v) for v in r[1:]]
        vals += [""] * (nz - len(vals))
        ws.append([r[0]] + vals)
        row = ws.max_row
        if r[0] in KEY_HELP:
            ws.cell(row, 1).comment = Comment(KEY_HELP[r[0]], APP_NAME)
        ws.cell(row, 1).font = Font(name=F, bold=True)
        ws.cell(row, 1).fill = PatternFill("solid", fgColor=keyfill)
        single = r[0] in ("titre_widget", "sousTitre_widget", "infos_widget")
        for col in range(2, nz + 2):
            cell = ws.cell(row, col)
            cell.font = Font(name=F, color="0000FF")
            cell.border = border
            if not (single and col > 2):
                cell.fill = PatternFill("solid", fgColor=inp)
        ws.cell(row, 1).border = border
    ws.freeze_panes = "B2"
    ws.column_dimensions["A"].width = 28
    for col in range(2, nz + 2):
        ws.column_dimensions[ws.cell(1, col).column_letter].width = 22

    info = wb.create_sheet("Mode d'emploi")
    info["A1"] = f"Modèle d'import — {APP_NAME} v{APP_VERSION}"
    info["A1"].font = Font(name=F, bold=True, size=14, color=dark)
    lines = [
        "Cellules jaunes (texte bleu) = saisies. Colonne A = clés : ne pas modifier.",
        "Colonne B = zone 1, colonne C = zone 2… Pour ajouter une zone, copier la dernière colonne de zone "
        "à sa droite et renommer la zone (ligne zone_widget). Ne rien saisir à droite des zones.",
        "Un commentaire sur certaines clés de la colonne A rappelle leur rôle.",
        "titre_widget, sousTitre_widget, infos_widget : paramètres généraux (colonne B uniquement).",
        "DH : jusqu'à 8 groupes. Pour chaque groupe : DH_Gi_widget (valeur), DH_Gi_nom_widget (nom affiché), "
        "DH_Gi_max_widget (DHmax propre au groupe). Laisser la valeur vide pour ne pas créer le groupe.",
        "DHmax_widget : DHmax par défaut, utilisé uniquement pour un groupe sans DHmax propre.",
        "Cibles (facultatives) : lignes *cible_widget (ex. Bbiocible_widget = 85 avec Bbiomax_widget = 120). "
        "La cible est tracée en pointillé gris, plus discrète que le max. Laisser vide pour ne pas l'afficher. "
        "DH : DH_Gi_cible_widget par groupe, DHcible_widget par défaut.",
        "show_* = 1 affiche l'indicateur, 0 le masque.",
        "Bâtiments : batiment_widget donne le bâtiment de chaque zone (un projet = 1 ou plusieurs bâtiments, "
        "un bâtiment = 1 ou plusieurs zones, une zone = 1 ou plusieurs groupes DH).",
        "Bilan bâtiment : Bbio, CEP, CEPnr, ICénergie et ICconstruction sont vérifiés au niveau bâtiment, "
        "le DH par groupe. Bâtiment à une zone : valeurs de la zone. Plusieurs zones : moyenne pondérée par "
        "surface_widget, ou mieux, saisir les résultats bâtiment Pléiades dans les lignes bat_*_widget "
        "(une seule colonne par bâtiment).",
        "Les lignes « ▸ … » sont des titres de section, ignorées à l'import.",
        "Import : bouton « Importer un fichier Excel » de l'application, ou copier tout le tableau et le coller "
        "dans la zone de collage.",
        "Valeurs pré-remplies = projet FICTIF de démonstration (2 bâtiments, zones tertiaire et commerce, "
        "groupes climatisé / non climatisé), à remplacer par les résultats du projet.",
    ]
    for i, t in enumerate(lines, 3):
        info.cell(i, 1, value=f"• {t}").font = Font(name=F)
        info.cell(i, 1).alignment = Alignment(wrap_text=True, vertical="top")
    info.column_dimensions["A"].width = 120
    buf = io.BytesIO()
    wb.save(buf)
    return buf.getvalue()


def excel_upload_to_text(data):
    wb = load_workbook(io.BytesIO(data), data_only=True, read_only=True)
    ws = wb["Saisie RE2020"] if "Saisie RE2020" in wb.sheetnames else wb[wb.sheetnames[0]]
    lines=[]
    for row in ws.iter_rows(values_only=True):
        vals=["" if v is None else str(v) for v in row]
        while vals and vals[-1]=="": vals.pop()
        if vals: lines.append("\t".join(vals))
    return "\n".join(lines)


_MIME = {".xlsx": "application/vnd.openxmlformats-officedocument.spreadsheetml.sheet",
         ".zip": "application/zip", ".json": "application/json", ".png": "image/png",
         ".svg": "image/svg+xml", ".pdf": "application/pdf"}


def download_link(data, filename, label):
    """Lien de téléchargement direct par le navigateur (fonctionne sur Binder : aucun
    fichier n'est lu sur le serveur, les octets sont intégrés au lien)."""
    mime = _MIME.get(os.path.splitext(filename)[1].lower(), "application/octet-stream")
    b64=base64.b64encode(data).decode("ascii")
    return f'<a download="{_html.escape(filename)}" href="data:{mime};base64,{b64}" style="display:inline-block;padding:7px 12px;background:#1F4E78;color:white;text-decoration:none;border-radius:4px">{_html.escape(label)}</a>'
```

```python
# ============================================================
# 6. INTERFACE UTILISATEUR (ipywidgets) — 4 étapes : Projet › Saisie › Graphiques › Export
# ============================================================
from collections import Counter

_STYLE = {"description_width": "170px"}
_COLORS = {"error": "#B3261E", "warning": "#8A5A00", "success": "#1E6B3A", "info": "#1F4E79"}
STEPS = [("projet", "1 · Projet"), ("saisie", "2 · Saisie"), ("graphs", "3 · Graphiques"), ("export", "4 · Export")]


def _sub(title, help_txt=""):
    """Sous-titre d'une étape."""
    h = f"<small style='color:#777'>&nbsp;— {help_txt}</small>" if help_txt else ""
    return W.HTML(f"<div style='margin:12px 0 4px;padding:3px 8px;border-left:4px solid #1F4E78;"
                  f"background:rgba(31,78,120,0.07)'><b>{title}</b>{h}</div>")


def _msg_html(kind, title, items=()):
    color = _COLORS.get(kind, "#333")
    li = "".join(f"<li>{_html.escape(str(i))}</li>" for i in items)
    body = f"<ul style='margin:4px 0 0 18px;padding:0'>{li}</ul>" if li else ""
    return (f"<div style='border-left:4px solid {color};padding:6px 10px;margin:4px 0;"
            f"background:rgba(127,127,127,0.06)'><b style='color:{color}'>"
            f"{_html.escape(title)}</b>{body}</div>")


def _banner_html(kind, text):
    bg = {"warning": "#FFF4D6", "success": "rgba(30,107,58,0.08)", "info": "rgba(31,78,121,0.08)"}.get(kind, "#eee")
    fg = _COLORS.get(kind, "#333")
    return (f"<div style='padding:8px 12px;border-radius:6px;background:{bg};color:{fg};"
            f"font-weight:bold'>{_html.escape(text)}</div>")


class IndicatorTab:
    """Onglet de saisie d'un indicateur (pour la zone affichée). Toutes les lignes existent
    toujours : masquer une ligne ne supprime pas sa saisie."""

    def __init__(self, key, on_change):
        self.key = key
        self.max = max_contrib(key)
        self._on_change = on_change
        self.include = W.Checkbox(value=True, description="Inclure dans les graphiques", indent=False)
        # Hors DH : nombre de contributions fixe (structure du modèle), pas de réglage.
        self.n = W.IntSlider(value=1 if key == "DH" else self.max, min=1, max=self.max, step=1,
                             description="Nombre de groupes DH", style=_STYLE,
                             layout=W.Layout(width="420px", display="" if key == "DH" else "none"))
        self.unit = W.Text(description="Unité", style=_STYLE, layout=W.Layout(width="420px"))
        self.per_group = key == "DH"
        self.target = W.Text(description=("DHmax par défaut" if self.per_group else "Valeur maximale (cible)"),
                             placeholder=("si groupe sans DHmax" if self.per_group else "ex. 63"),
                             style=_STYLE, layout=W.Layout(width="420px"))
        self.goal = W.Text(description=("Cible DH par défaut" if self.per_group else "Cible (facultative)"),
                           placeholder=("si groupe sans cible" if self.per_group else "ex. 85 — vide = aucune"),
                           style=_STYLE, layout=W.Layout(width="420px"))
        self.keys, self.key_html = [""] * self.max, []
        self.labels, self.values, self.colors, self.rows, self.maxes, self.goals = [], [], [], [], [], []
        header = W.HBox([
            W.HTML("<b>N°</b>", layout=W.Layout(width="34px")),
            W.HTML("<b>Clé Excel</b>", layout=W.Layout(width="140px")),
            W.HTML("<b>Libellé</b>", layout=W.Layout(width="340px")),
            W.HTML("<b>Valeur</b>", layout=W.Layout(width="110px")),
            *([W.HTML("<b>DHmax du groupe</b>", layout=W.Layout(width="130px")),
               W.HTML("<b>Cible du groupe</b>", layout=W.Layout(width="130px"))] if self.per_group else []),
            W.HTML("<b>Couleur</b>", layout=W.Layout(width="180px")),
        ])
        for i in range(self.max):
            kh = W.HTML(layout=W.Layout(width="140px"))
            lab = W.Text(placeholder=f"Contribution {i + 1}", layout=W.Layout(width="340px"))
            val = W.Text(value="0", layout=W.Layout(width="110px"))
            col = W.ColorPicker(value=PALETTE[i % len(PALETTE)], concise=False, layout=W.Layout(width="180px"))
            mx = W.Text(value="", placeholder="DHmax", layout=W.Layout(width="130px"))
            gl = W.Text(value="", placeholder="cible", layout=W.Layout(width="130px"))
            self.maxes.append(mx); self.goals.append(gl)
            row = W.HBox([W.HTML(f"{i + 1}", layout=W.Layout(width="34px")), kh, lab, val,
                          *([mx, gl] if self.per_group else []), col])
            self.key_html.append(kh); self.labels.append(lab); self.values.append(val)
            self.colors.append(col); self.rows.append(row)
        self.total_html = W.HTML()
        self.msg_html = W.HTML()
        self.widget = W.VBox([
            self.include, self.n, self.unit, self.target, self.goal,
            W.HTML("<hr style='margin:6px 0'>"), header, *self.rows,
            W.HTML("<hr style='margin:6px 0'>"), self.total_html, self.msg_html,
        ], layout=W.Layout(padding="6px"))

        for w in [self.include, self.n, self.unit, self.target, self.goal, *self.labels, *self.values,
                  *self.maxes, *self.goals, *self.colors]:
            w.observe(self._changed, names="value")
        self._update_rows()

    def _changed(self, change):
        if change["owner"] is self.n:
            self._update_rows()
        self._on_change(self)

    def _update_rows(self):
        if not self.per_group:
            self.n.value = self.max          # hors DH : toutes les contributions du modèle
        for i, row in enumerate(self.rows):
            row.layout.display = "" if i < self.n.value else "none"

    def _show_key(self, i):
        k = self.keys[i]
        self.key_html[i].value = (f"<code style='font-size:11px'>{_html.escape(excel_name(k))}</code>"
                                  if k else "<span style='color:#999'>—</span>")

    def get_raw(self):
        return {
            "unit": self.unit.value,
            "target": self.target.value,
            "goal": self.goal.value,
            "n_active": int(self.n.value),
            "include": bool(self.include.value),
            "contributions": [{"key": self.keys[i], "label": self.labels[i].value,
                               "value": self.values[i].value, "max": self.maxes[i].value,
                               "goal": self.goals[i].value,
                               "color": self.colors[i].value}
                              for i in range(self.max)],
        }

    def set_raw(self, raw):
        self.include.value = bool(raw.get("include", True))
        self.n.value = min(max(int(raw.get("n_active", 1)), 1), self.max) if self.per_group else self.max
        self.unit.value = str(raw.get("unit", ""))
        self.target.value = str(raw.get("target", ""))
        self.goal.value = str(raw.get("goal", "") or "")
        for i, c in enumerate(raw["contributions"][:self.max]):
            self.keys[i] = str(c.get("key", "") or "")
            self._show_key(i)
            self.labels[i].value = str(c.get("label", ""))
            self.values[i].value = str(c.get("value", ""))
            self.maxes[i].value = str(c.get("max", "") or "")
            self.goals[i].value = str(c.get("goal", "") or "")
            color, _ = normalize_color(c.get("color"), PALETTE[i % len(PALETTE)])
            self.colors[i].value = color
        self._update_rows()

    def refresh_feedback(self, s):
        """Met à jour l'aperçu du total et les messages de l'onglet. Retourne la validation."""
        res = validate_indicator(self.key, self.get_raw())
        if not self.include.value:
            self.total_html.value = "<i>Indicateur exclu des graphiques pour cette zone.</i>"
            self.msg_html.value = ""
            return res
        total, ok_vals = 0.0, True
        for i in range(self.n.value):
            v, err = parse_number(self.values[i].value)
            if err or v < 0:
                ok_vals = False
            else:
                total += v
        tgt, terr = parse_number(self.target.value)
        unit = _html.escape(self.unit.value.strip())
        if self.per_group and ok_vals:
            parts = []
            for i in range(self.n.value):
                v, _ = parse_number(self.values[i].value)
                m, merr = parse_number(self.maxes[i].value)
                if merr and terr is None:
                    m, merr = tgt, None
                lab = _html.escape(self.labels[i].value.strip() or f"Groupe {i + 1}")
                if merr is None and m > 0:
                    parts.append(f"{lab} : {fmt_num(v, s.decimals, s.decimal_sep)} / "
                                 f"{fmt_target(m, s.decimals, s.decimal_sep)} {unit} "
                                 f"({fmt_num(100 * v / m, 0, s.decimal_sep)} % du max)")
                    g, gerr = parse_number(self.goals[i].value)
                    if gerr:
                        g, gerr = parse_number(self.goal.value)
                    if gerr is None and g > 0:
                        parts[-1] += f" — cible {fmt_target(g, s.decimals, s.decimal_sep)}"
                else:
                    parts.append(f"{lab} : {fmt_num(v, s.decimals, s.decimal_sep)} {unit} (DHmax manquant)")
            self.total_html.value = "<b>Groupes DH</b><br>" + "<br>".join(parts)
            ok_vals = None
        if ok_vals is None:
            pass
        elif ok_vals:
            txt = f"<b>Total calculé : {fmt_num(total, s.decimals, s.decimal_sep)} {unit}</b>"
            if terr is None and tgt > 0:
                txt += (f" &nbsp;|&nbsp; Max : {fmt_target(tgt, s.decimals, s.decimal_sep)} {unit}"
                        f" &nbsp;|&nbsp; {fmt_num(100 * total / tgt, 0, s.decimal_sep)} % du max")
            g, gerr = parse_number(self.goal.value)
            if gerr is None and g > 0:
                txt += (f" &nbsp;|&nbsp; Cible : {fmt_target(g, s.decimals, s.decimal_sep)} {unit}"
                        f" ({fmt_num(100 * total / g, 0, s.decimal_sep)} %)")
        else:
            txt = "<b>Total calculé : —</b> (valeurs à corriger)"
        if ok_vals is not None:
            self.total_html.value = txt
        if res.errors:
            self.msg_html.value = _msg_html("error", "À corriger", res.errors)
        elif res.warnings:
            self.msg_html.value = _msg_html("warning", "Remarques", res.warnings)
        else:
            self.msg_html.value = _msg_html("success", "Saisie valide")
        return res


class RE2020App:
    """Interface en 4 étapes : 1. Projet (structure bâtiments › zones, titres)  2. Saisie (import
    Excel, saisies à corriger, saisie manuelle)  3. Graphiques (familles cochées, actualisation
    manuelle, un aperçu à la fois)  4. Export. Seules les familles cochées sont calculées ;
    l'export calcule à la demande les familles demandées."""

    def __init__(self, state=None):
        self._loading = False
        self._loading_prev = False
        self.buildings = [new_building("Bâtiment 1")]
        self.zones = [new_zone("Zone 1")]
        self.cur = 0          # zone sélectionnée
        self.curb = 0         # bâtiment sélectionné
        self._cache = {}      # famille -> [(légende, png, largeur px)] (aperçus déjà calculés)
        self._data_dirty = True
        self._counts = {}     # famille -> nombre de figures
        self._previews = []
        self._nav_sig = None
        st = normalize_state(state or default_state())
        L = W.Layout

        # ======================== barre d'étapes
        self.w_step = W.ToggleButtons(options=[(lab, k) for k, lab in STEPS], value="projet",
                                      style={"button_width": "150px", "font_weight": "bold"})

        # ======================== 1. PROJET
        self.btn_new = W.Button(description="Projet vierge", icon="file", layout=L(width="140px"))
        self.btn_demo = W.Button(description="Projet exemple", icon="flask", layout=L(width="150px"))
        self.w_json_upload = W.FileUpload(accept=".json", multiple=False, description="Choisir un JSON",
                                          layout=L(width="160px"))
        self.btn_load_upload = W.Button(description="Ouvrir", icon="folder-open", layout=L(width="100px"))
        self.w_json = W.Text(value="projet_RE2020.json", placeholder="nom du fichier .json",
                             layout=L(width="220px"))
        self.btn_save = W.Button(description="Sauvegarder", icon="save", button_style="info",
                                 layout=L(width="140px"))
        self.btn_load = W.Button(description="Charger (chemin local)", icon="folder-open",
                                 layout=L(width="200px"))
        project_bar = W.VBox([
            W.HBox([self.btn_new, self.btn_demo,
                    W.HTML("<span style='color:#bbb'>&nbsp;|&nbsp;</span>"),
                    self.w_json_upload, self.btn_load_upload,
                    W.HTML("<span style='color:#bbb'>&nbsp;|&nbsp;</span>"),
                    self.w_json, self.btn_save] + ([] if IN_BROWSER else [self.btn_load]),
                   layout=L(flex_flow="row wrap", align_items="center")),
            W.HTML("<small>Ouvrir : choisir le fichier JSON puis « Ouvrir ». Sauvegarder : un bouton de "
                   "téléchargement apparaît en bas de l'outil.</small>"),
        ])

        # --- arbre bâtiments › zones
        self.tree_box = W.VBox(layout=L(padding="2px 0"))
        self.btn_bld_add = W.Button(description="Bâtiment", icon="plus", tooltip="Ajouter un bâtiment (avec une zone vide)",
                                    layout=L(width="110px"))
        self.btn_zone_add = W.Button(description="Zone", icon="plus", tooltip="Ajouter une zone au bâtiment sélectionné",
                                     layout=L(width="90px"))
        self.btn_zone_dup = W.Button(description="Dupliquer", icon="copy", tooltip="Dupliquer la zone sélectionnée",
                                     layout=L(width="110px"))
        self.btn_zone_del = W.Button(description="Zone", icon="minus", button_style="warning",
                                     tooltip="Supprimer la zone sélectionnée", layout=L(width="90px"))
        self.btn_bld_del = W.Button(description="Bâtiment", icon="minus", button_style="warning",
                                    tooltip="Supprimer le bâtiment sélectionné et ses zones", layout=L(width="110px"))
        tree_col = W.VBox([
            W.HTML("<b>Bâtiments › zones</b> <small style='color:#777'>(cliquer pour sélectionner)</small>"),
            self.tree_box,
            W.HBox([self.btn_bld_add, self.btn_zone_add, self.btn_zone_dup], layout=L(flex_flow="row wrap")),
            W.HBox([self.btn_zone_del, self.btn_bld_del], layout=L(flex_flow="row wrap")),
        ], layout=L(flex="0 1 340px", min_width="260px", border="1px solid #c8c8c8", padding="8px",
                    margin="0 12px 8px 0"))
        self.w_bld_name = W.Text(description="Nom", style={"description_width": "90px"}, layout=L(width="320px"))
        self.w_zone_name = W.Text(description="Nom", style={"description_width": "90px"}, layout=L(width="320px"))
        self.w_zone_surf = W.Text(description="Surface (m²)", placeholder="pondération du bilan",
                                  style={"description_width": "90px"}, layout=L(width="320px"))
        self.w_zone_bld = W.Dropdown(description="Bâtiment", style={"description_width": "90px"},
                                     layout=L(width="320px"))
        self.w_detail_hint = W.HTML()
        detail_col = W.VBox([
            W.HTML("<b>🏢 Bâtiment sélectionné</b>"), self.w_bld_name,
            W.HTML("<b style='display:block;margin-top:8px'>📐 Zone sélectionnée</b>"),
            self.w_zone_name, self.w_zone_surf, self.w_zone_bld, self.w_detail_hint,
        ], layout=L(flex="1 1 340px", border="1px solid #c8c8c8", padding="8px", margin="0 0 8px 0"))
        self.w_title = W.Text(description="Titre", style={"description_width": "90px"},
                              layout=L(width="100%", max_width="760px"))
        self.w_subtitle = W.Text(description="Sous-titre", style={"description_width": "90px"},
                                 layout=L(width="100%", max_width="760px"))
        self.w_footer = W.Textarea(description="Infos en bas", style={"description_width": "90px"},
                                   layout=L(width="100%", max_width="760px", height="60px"))
        self.btn_next1 = W.Button(description="Suivant : saisie", icon="arrow-right", button_style="primary",
                                  layout=L(width="190px", margin="10px 0 0 0"))
        step1 = W.VBox([
            _sub("Projet", "partir d'un projet vierge, de l'exemple, ou rouvrir un JSON"), project_bar,
            _sub("Structure", "projet → bâtiments → zones (→ groupes DH, à l'étape 2)"),
            W.HBox([tree_col, detail_col], layout=L(flex_flow="row wrap", align_items="flex-start")),
            _sub("Titres des graphiques"), self.w_title, self.w_subtitle, self.w_footer,
            self.btn_next1,
        ])

        # ======================== 2. SAISIE
        self.w_excel_template = W.HTML(value=download_link(excel_template_bytes(demo_state()),
                                                           "Modele_import_RE2020.xlsx",
                                                           "Télécharger le modèle Excel"))
        self.w_excel_upload = W.FileUpload(accept=".xlsx", multiple=False, description="Choisir un Excel")
        self.btn_excel_import = W.Button(description="Importer le fichier", icon="upload",
                                         button_style="success", layout=L(width="200px"))
        self.w_paste = W.Textarea(placeholder="Collez ici des lignes du tableau Excel (clé | valeur zone 1 | "
                                              "valeur zone 2 …)\nex. BbioChaud_widget\t48,8",
                                  layout=L(width="100%", height="150px"))
        self.w_paste_zone = W.Dropdown(description="1re colonne de valeurs →", style=_STYLE,
                                       layout=L(width="480px"))
        self.btn_paste = W.Button(description="Remplir à partir du collage", icon="paste",
                                  button_style="success", layout=L(width="240px"))
        self.btn_paste_clear = W.Button(description="Vider", icon="eraser", layout=L(width="100px"))
        self.btn_to_excel = W.Button(description="Générer le tableau des valeurs actuelles", icon="table",
                                     layout=L(width="320px"))
        self.w_excel_out = W.Textarea(layout=L(width="100%", height="130px"),
                                      placeholder="Le tableau à copier dans Excel apparaîtra ici.")
        paste_box = W.VBox([
            W.HTML("<small>Copiez tout ou partie du tableau (colonne des clés + colonnes de valeurs). "
                   "Seules les valeurs collées sont modifiées.</small>"),
            self.w_paste, self.w_paste_zone, W.HBox([self.btn_paste, self.btn_paste_clear]),
            W.HTML("<hr style='margin:8px 0'><b>Renvoyer les valeurs vers Excel</b> <small>: générez le "
                   "tableau, Ctrl+A / Ctrl+C dedans, puis collez en A1 dans Excel.</small>"),
            self.btn_to_excel, self.w_excel_out,
        ])
        self.paste_acc = W.Accordion(children=[paste_box])
        self.paste_acc.set_title(0, "Copier-coller depuis Excel / renvoyer les valeurs vers Excel")
        self.paste_acc.selected_index = None

        self.err_box = W.VBox()

        self.w_zone_sel = W.Dropdown(description="Zone en cours", style={"description_width": "95px"},
                                     layout=L(width="100%", max_width="520px"))
        self.crumb_top = W.HTML()
        self.crumb_bottom = W.HTML()
        self.w_bld_mode = W.ToggleButtons(options=[("Calculés depuis les zones", "auto"),
                                                   ("Saisis directement (Pléiades)", "manual")],
                                          description="Résultats bâtiment", style=_STYLE)
        self.bld_val, self.bld_max = {}, {}
        rows = [W.HBox([W.HTML("<b>Indicateur</b>", layout=L(width="150px")),
                        W.HTML("<b>Valeur bâtiment</b>", layout=L(width="140px")),
                        W.HTML("<b>Seuil max</b>", layout=L(width="140px"))])]
        for k in BUILDING_KEYS:
            self.bld_val[k] = W.Text(layout=L(width="130px"), placeholder="valeur")
            self.bld_max[k] = W.Text(layout=L(width="130px"), placeholder="max")
            rows.append(W.HBox([W.HTML(f"{k} <small style='color:#777'>({DEFAULT_UNITS[k]})</small>",
                                       layout=L(width="150px")), self.bld_val[k], self.bld_max[k]]))
        self.bld_manual = W.VBox(rows)
        self.bld_info = W.HTML()
        self.bld_results = W.Accordion(children=[W.VBox([self.w_bld_mode, self.bld_manual, self.bld_info])])
        self.bld_results.set_title(0, "Résultats au niveau bâtiment")
        self.bld_results.selected_index = None
        self.tabs = {k: IndicatorTab(k, self._on_tab_change) for k in INDICATOR_KEYS}
        self.tab_widget = W.Tab(children=[t.widget for t in self.tabs.values()])
        for i, k in enumerate(INDICATOR_KEYS):
            self.tab_widget.set_title(i, k)
        manual_box = W.VBox([self.w_zone_sel, self.crumb_top, self.bld_results, self.tab_widget, self.crumb_bottom])
        self.manual_acc = W.Accordion(children=[manual_box])
        self.manual_acc.set_title(0, "Saisie manuelle et contrôle des valeurs (zone par zone)")
        self.manual_acc.selected_index = None
        self.btn_prev2 = W.Button(description="Projet", icon="arrow-left", layout=L(width="120px"))
        self.btn_next2 = W.Button(description="Suivant : graphiques", icon="arrow-right", button_style="primary",
                                  layout=L(width="210px"))
        step2 = W.VBox([
            _sub("A. Importer depuis Excel", "recommandé : une colonne par zone, ligne batiment_widget = bâtiment"),
            W.HBox([self.w_excel_template, W.HTML("&nbsp;&nbsp;"), self.w_excel_upload, self.btn_excel_import],
                   layout=L(flex_flow="row wrap", align_items="center")),
            self.paste_acc,
            _sub("B. Saisies à corriger", "cliquer pour aller directement au champ"),
            self.err_box,
            _sub("C. Saisie manuelle", "pour compléter ou vérifier, zone par zone"),
            self.manual_acc,
            W.HBox([self.btn_prev2, self.btn_next2], layout=L(margin="10px 0 0 0")),
        ])

        # ======================== 3. GRAPHIQUES
        self.chips = {}
        for k, lab, tip in FAMILIES:
            t = W.ToggleButton(value=k in DEFAULT_FAMILIES, description=lab, tooltip=tip,
                               icon="check" if k in DEFAULT_FAMILIES else "",
                               layout=L(width="auto", min_width="130px", margin="2px 6px 2px 0"))
            t.observe(self._on_family, names="value")
            self.chips[k] = t
        self.btn_preset_synth = W.Button(description="Note de synthèse", icon="file-o",
                                         tooltip="Synthèse seule", layout=L(width="auto"))
        self.btn_preset_all = W.Button(description="Rapport complet", icon="files-o",
                                       tooltip="Toutes les familles", layout=L(width="auto"))
        self.w_banner = W.HTML(layout=L(flex="1 1 300px"))
        self.btn_refresh = W.Button(description="Actualiser", icon="refresh", button_style="primary",
                                    layout=L(width="150px", height="36px"))
        self.btn_prev = W.Button(icon="chevron-left", tooltip="Graphique précédent", layout=L(width="44px"))
        self.btn_next = W.Button(icon="chevron-right", tooltip="Graphique suivant", layout=L(width="44px"))
        self.w_prev_sel = W.Dropdown(layout=L(width="auto", flex="1 1 280px", max_width="600px"))
        self.w_prev_pos = W.HTML(layout=L(width="70px"))
        self.w_img = W.Image(format="png", layout=L(max_width="100%", height="auto", border="1px solid #e0e0e0",
                                                    display="none"))
        self.w_prev_msg = W.HTML()
        self.w_prev_sel.observe(self._on_prev_select, names="value")

        self.w_brand = W.Checkbox(value=True, indent=False,
                                  description="Charte graphique WSP (décocher = style de base)")
        self.w_width = W.BoundedFloatText(min=8, max=60, step=0.5, description="Largeur (cm)",
                                          style=_STYLE, layout=L(width="300px"))
        self.w_h_over = W.BoundedFloatText(min=0, max=100, step=0.5, description="Hauteur ensemble (cm)",
                                           style=_STYLE, layout=L(width="300px"))
        self.w_h_single = W.BoundedFloatText(min=0, max=100, step=0.5, description="Hauteur individuelle (cm)",
                                             style=_STYLE, layout=L(width="300px"))
        self.w_font = W.IntSlider(min=6, max=20, description="Taille des textes (pt)", style=_STYLE,
                                  layout=L(width="420px"))
        self.w_dec = W.IntSlider(min=0, max=4, description="Nombre de décimales", style=_STYLE,
                                 layout=L(width="420px"))
        self.w_sep = W.ToggleButtons(options=[("Virgule", ","), ("Point", ".")],
                                     description="Séparateur décimal", style=_STYLE)
        self.w_show = W.Checkbox(description="Afficher les valeurs dans les segments", indent=False)
        self.w_hide0 = W.Checkbox(description="Masquer les contributions nulles dans la légende", indent=False)
        self.w_common = W.Checkbox(description="Échelle commune entre zones (vues par indicateur)", indent=False)
        format_box = W.VBox([
            self.w_brand,
            W.HTML("<small>Charte WSP : bandeau noir, touches de rouge WSP, titres en capitales, palette "
                   "data WSP. Les couleurs modifiées à la main sont conservées.</small>"),
            W.HBox([self.w_width, self.w_h_over, self.w_h_single], layout=L(flex_flow="row wrap")),
            W.HTML("<small>Hauteur 0 = automatique (recommandé). Synthèse et bâtiment(s) sont toujours en "
                   "hauteur automatique.</small>"),
            self.w_font, self.w_dec, self.w_sep, self.w_show, self.w_hide0, self.w_common,
        ])
        self.adv_acc = W.Accordion(children=[format_box])
        self.adv_acc.set_title(0, "Réglages avancés (charte, dimensions, textes)")
        self.adv_acc.selected_index = None
        self.btn_prev3 = W.Button(description="Saisie", icon="arrow-left", layout=L(width="120px"))
        self.btn_next3 = W.Button(description="Suivant : export", icon="arrow-right", button_style="primary",
                                  layout=L(width="190px"))
        step3 = W.VBox([
            _sub("Graphiques à afficher", "seules les familles cochées sont calculées"),
            W.HBox(list(self.chips.values()), layout=L(flex_flow="row wrap")),
            W.HBox([W.HTML("<small style='color:#777'>Préréglages :&nbsp;</small>"),
                    self.btn_preset_synth, self.btn_preset_all],
                   layout=L(flex_flow="row wrap", align_items="center", margin="4px 0")),
            W.HBox([self.w_banner, self.btn_refresh],
                   layout=L(flex_flow="row wrap", align_items="center", margin="6px 0")),
            _sub("Aperçu", "un graphique à la fois, résolution réduite"),
            W.HBox([self.btn_prev, self.w_prev_sel, self.btn_next, self.w_prev_pos],
                   layout=L(flex_flow="row wrap", align_items="center")),
            self.w_prev_msg, self.w_img,
            self.adv_acc,
            W.HBox([self.btn_prev3, self.btn_next3], layout=L(margin="10px 0 0 0")),
        ])

        # ======================== 4. EXPORT
        self.w_exp_what = W.Dropdown(options=[("Graphiques cochés à l'étape 3", "checked"),
                                              ("Tous les graphiques", "all")]
                                     + [(lab, k) for k, lab, _ in FAMILIES],
                                     description="Exporter", style={"description_width": "70px"},
                                     layout=L(width="360px"))
        self.w_png = W.Checkbox(description="PNG", indent=False, layout=L(width="70px"))
        self.w_svg = W.Checkbox(description="SVG", indent=False, layout=L(width="70px"))
        self.w_pdf = W.Checkbox(description="PDF", indent=False, layout=L(width="70px"))
        self.w_dpi = W.BoundedIntText(min=300, max=1200, step=50, description="PNG (dpi)",
                                      style={"description_width": "70px"}, layout=L(width="170px"))
        self.w_bg = W.ToggleButtons(options=[("Fond blanc", "blanc"), ("Transparent", "transparent")],
                                    style={"button_width": "100px"})
        self.w_prefix = W.Text(description="Préfixe", style={"description_width": "70px"}, layout=L(width="260px"))
        self.w_dir = W.Text(description="Dossier", style={"description_width": "60px"}, layout=L(width="300px"))
        self.w_disk = W.Checkbox(value=not REMOTE, indent=False,
                                 description="Écrire aussi dans le dossier (usage local)")
        self.btn_export = W.Button(description="Exporter (ZIP)", icon="download", button_style="primary",
                                   layout=L(width="180px", height="36px"))
        self.btn_prev4 = W.Button(description="Graphiques", icon="arrow-left", layout=L(width="140px"))
        step4 = W.VBox([
            _sub("Quoi exporter", "les graphiques sont recalculés en haute résolution au moment de l'export"),
            self.w_exp_what,
            W.HBox([W.HTML("<b>Formats :</b>&nbsp;"), self.w_png, self.w_svg, self.w_pdf, self.w_dpi],
                   layout=L(flex_flow="row wrap", align_items="center")),
            W.HBox([self.w_bg, self.w_prefix] + ([] if IN_BROWSER else [self.w_dir, self.w_disk]),
                   layout=L(flex_flow="row wrap", align_items="center")),
            self.btn_export,
            W.HTML("<small>Exports nombreux (toutes les vues, PNG 300 dpi + SVG + PDF) : patienter jusqu'à "
                   "l'apparition du bouton de téléchargement.</small>"),
            self.btn_prev4,
        ])

        # ======================== assemblage
        self.status = W.HTML()
        self.dl_box = W.HTML()
        self.panels = {"projet": step1, "saisie": step2, "graphs": step3, "export": step4}
        self.widget = W.VBox([
            W.HTML(f"<h3 style='margin:4px 0'>Graphiques RE2020 — bullet graphs et bilan bâtiment "
                   f"<small style='color:#777'>v{APP_VERSION}</small></h3>"
                   + ("<small style='color:#777'>Tout est calculé dans votre navigateur : "
                      "aucune donnée n'est envoyée sur un serveur.</small>" if IN_BROWSER else "")),
            self.w_step,
            W.HTML("<div style='border-top:2px solid #1F4E78;margin:2px 0 0'></div>"),
            *[W.VBox([p]) for p in self.panels.values()],
            self.status, self.dl_box,
        ])
        self._panel_boxes = dict(zip(self.panels, self.widget.children[3:3 + len(self.panels)]))

        actions = [(self.btn_refresh, self.generate), (self.btn_export, self.export_selected),
                   (self.btn_save, self.save), (self.btn_load, self.load), (self.btn_demo, self.load_demo),
                   (self.btn_new, self.new_project), (self.btn_paste, self.paste_import),
                   (self.btn_paste_clear, self.paste_clear), (self.btn_to_excel, self.to_excel),
                   (self.btn_excel_import, self.import_excel_file), (self.btn_load_upload, self.load_upload),
                   (self.btn_zone_add, self.zone_add), (self.btn_zone_dup, self.zone_duplicate),
                   (self.btn_zone_del, self.zone_delete), (self.btn_bld_add, self.building_add),
                   (self.btn_bld_del, self.building_delete),
                   (self.btn_preset_synth, lambda: self._preset({"synth"})),
                   (self.btn_preset_all, lambda: self._preset(set(FAMILY_KEYS)))]
        self._buttons = [b for b, _ in actions]
        for b, f in actions:
            b.on_click(self._safe(f))
        for b, step in ((self.btn_next1, "saisie"), (self.btn_prev2, "projet"), (self.btn_next2, "graphs"),
                        (self.btn_prev3, "saisie"), (self.btn_next3, "export"), (self.btn_prev4, "graphs")):
            b.on_click(lambda _, step=step: setattr(self.w_step, "value", step))
        self.btn_prev.on_click(lambda _: self._step_preview(-1))
        self.btn_next.on_click(lambda _: self._step_preview(+1))
        self.w_step.observe(self._on_step, names="value")
        # réglages qui changent le rendu : graphiques à actualiser
        for w in [self.w_title, self.w_subtitle, self.w_footer, self.w_width, self.w_h_over, self.w_h_single,
                  self.w_font, self.w_show, self.w_hide0, self.w_common]:
            w.observe(lambda ch: self._mark_dirty(), names="value")
        for w in [self.w_dec, self.w_sep]:
            w.observe(lambda ch: (self._refresh_all_feedback(), self._mark_dirty()), names="value")
        self.w_zone_sel.observe(self._on_zone_select, names="value")
        self.w_zone_name.observe(self._on_zone_rename, names="value")
        self.w_zone_bld.observe(self._on_zone_building, names="value")
        self.w_zone_surf.observe(self._on_zone_surface, names="value")
        self.w_bld_name.observe(self._on_building_rename, names="value")
        self.w_bld_mode.observe(self._on_building_mode, names="value")
        for w in list(self.bld_val.values()) + list(self.bld_max.values()):
            w.observe(self._on_building_values, names="value")
        self.w_brand.observe(self._on_brand_change, names="value")

        self.set_state(st)
        self._show_step("projet")

    # ---------- utilitaires
    def _safe(self, func):
        @functools.wraps(func)
        def wrapper(*_):
            for b in self._buttons:
                b.disabled = True
            try:
                func()
            except Exception as e:   # message lisible plutôt qu'une trace d'erreur
                self._status("error", "Erreur inattendue", [f"{type(e).__name__} : {e}"])
            finally:
                for b in self._buttons:
                    b.disabled = False
                self._update_delete_buttons()
                self._refresh_banner()
        return wrapper

    def _update_delete_buttons(self):
        self.btn_zone_del.disabled = len(self._zones_in(self.curb)) < 2
        self.btn_bld_del.disabled = len(self.buildings) < 2

    def _status(self, kind, title, items=()):
        self.status.value = _msg_html(kind, title, items)

    def _settings_raw(self):
        return {
            "title": self.w_title.value, "subtitle": self.w_subtitle.value, "footer": self.w_footer.value,
            "width_cm": self.w_width.value,
            "height_overview_cm": self.w_h_over.value, "height_single_cm": self.w_h_single.value,
            "font_size": self.w_font.value, "show_values": self.w_show.value,
            "hide_zero_legend": self.w_hide0.value, "common_scale": self.w_common.value,
            "decimals": self.w_dec.value, "background": self.w_bg.value,
            "decimal_sep": self.w_sep.value, "dpi": self.w_dpi.value, "brand": self.w_brand.value,
            "export_dir": self.w_dir.value, "file_prefix": self.w_prefix.value,
            "formats": [f for f, w in (("png", self.w_png), ("svg", self.w_svg), ("pdf", self.w_pdf)) if w.value],
        }

    def _zones_in(self, bi):
        return [i for i, z in enumerate(self.zones) if int(z.get("building", 0)) == bi]

    def _bname(self, z):
        return self.buildings[min(int(z.get("building", 0)), len(self.buildings) - 1)]["name"]

    def _zone_has_errors(self, z):
        return any(raw["include"] and not validate_indicator(k, raw).ok
                   for k, raw in z["indicators"].items())

    # ---------- étapes
    def _show_step(self, key):
        for k, box in self._panel_boxes.items():
            box.layout.display = "" if k == key else "none"

    def _on_step(self, change):
        self._show_step(change["new"])
        if change["new"] == "graphs":
            self._safe(self._update_counts)()

    # ---------- navigation : arbre (étape 1), liste des zones et saisies à corriger (étape 2)
    def _refresh_nav(self, force=False):
        """Met à jour l'arbre, la liste des zones et la liste des saisies à corriger.
        Les widgets ne sont reconstruits que si quelque chose a changé (noms, erreurs, sélection)."""
        errs = [(zi, k) for zi, z in enumerate(self.zones) for k, raw in z["indicators"].items()
                if raw["include"] and not validate_indicator(k, raw).ok]
        sig = (tuple(b["name"] for b in self.buildings),
               tuple((z["name"], int(z.get("building", 0))) for z in self.zones),
               tuple(errs), self.cur, self.curb)
        if sig == self._nav_sig and not force:
            return
        self._nav_sig = sig
        bad_z = {zi for zi, _ in errs}
        multi_b = len(self.buildings) > 1
        L = W.Layout
        # arbre
        rows = []
        for bi, b in enumerate(self.buildings):
            zs = self._zones_in(bi)
            bad = any(zi in bad_z for zi in zs)
            bt = W.Button(description=f"🏢 {b['name']}" + (" ⚠" if bad else ""),
                          tooltip=f"{len(zs)} zone(s)", button_style="primary" if bi == self.curb else "",
                          layout=L(width="auto", min_width="220px", margin="6px 0 1px 0"))
            bt.on_click(self._safe(functools.partial(self._select_building, bi)))
            rows.append(bt)
            for zi in zs:
                z = self.zones[zi]
                zb = W.Button(description=f"📐 {z['name']}" + (" ⚠" if zi in bad_z else ""),
                              tooltip=f"Surface : {z.get('surface') or '—'} m²",
                              button_style="info" if zi == self.cur else "",
                              layout=L(width="auto", min_width="190px", margin="1px 0 1px 26px"))
                zb.on_click(self._safe(functools.partial(self._goto, zi)))
                rows.append(zb)
        self.tree_box.children = rows
        # listes de zones
        self._loading = True
        try:
            lab = lambda i, z: (f"{self._bname(z)} › " if multi_b else "") + z["name"] + (" ⚠" if i in bad_z else "")
            self.w_zone_sel.options = [(lab(i, z), i) for i, z in enumerate(self.zones)]
            self.w_zone_sel.value = self.cur
            prev = self.w_paste_zone.value
            self.w_paste_zone.options = ([(f"{i + 1}. " + lab(i, z).replace(" ⚠", ""), i)
                                          for i, z in enumerate(self.zones)] + [("Nouvelle zone", -1)])
            valid = [o[1] for o in self.w_paste_zone.options]
            self.w_paste_zone.value = prev if prev in valid else self.cur
            self.w_zone_bld.options = [(b["name"], i) for i, b in enumerate(self.buildings)]
            self.w_zone_bld.value = int(self.zones[self.cur].get("building", 0))
        finally:
            self._loading = False
        # saisies à corriger
        if not errs:
            self.err_box.children = [W.HTML(_msg_html("success", "Aucune saisie à corriger."))]
        else:
            btns = []
            for zi, k in errs[:30]:
                z = self.zones[zi]
                b = W.Button(description=(f"⚠ {self._bname(z)} › " if multi_b else "⚠ ") + f"{z['name']} — {k}",
                             button_style="warning", layout=L(width="auto", margin="2px 6px 2px 0"))
                b.on_click(self._safe(functools.partial(self._goto_error, zi, k)))
                btns.append(b)
            more = [W.HTML(f"<small>… et {len(errs) - 30} autre(s).</small>")] if len(errs) > 30 else []
            self.err_box.children = [W.HTML(_msg_html("error", f"{len(errs)} saisie(s) à corriger")),
                                     W.HBox(btns, layout=L(flex_flow="row wrap")), *more]
        self._update_delete_buttons()

    def _goto_error(self, zi, k):
        self._goto(zi)
        self.manual_acc.selected_index = 0
        self.tab_widget.selected_index = INDICATOR_KEYS.index(k)
        self.w_step.value = "saisie"

    # ---------- rappel de la position de saisie
    def _refresh_crumb(self):
        b = self.buildings[self.curb]
        z = self.zones[self.cur]
        a, _ = parse_number(z.get("surface", ""))
        surf = f" &nbsp;·&nbsp; {fmt_num(a, 0, ',')} m²" if a else ""
        raw = z["indicators"]["DH"]
        ng = raw["n_active"] if raw["include"] else 0
        e = _html.escape
        self.crumb_top.value = (
            "<div style='background:#111;color:#fff;padding:9px 14px;margin:4px 0;border-left:8px solid "
            "#FF372F;font-size:15px'>✎ <span style='color:#bbb;font-size:12px;letter-spacing:1px'>SAISIE EN "
            f"COURS</span>&nbsp;&nbsp;<b>{e(b['name'])}</b> &nbsp;›&nbsp; <b style='color:#FFD10A'>"
            f"{e(z['name'])}</b><span style='color:#bbb;font-size:12px'>{surf} &nbsp;·&nbsp; {ng} groupe(s) DH"
            "</span></div>")
        self.crumb_bottom.value = (
            "<div style='color:#555;font-size:12px;padding:3px 8px;border-top:1px solid #ddd'>↑ Valeurs saisies "
            f"pour <b>{e(b['name'])} › {e(z['name'])}</b></div>")
        self.bld_results.set_title(0, f"Résultats au niveau bâtiment — {b['name']} : "
                                      + ("saisis directement" if b.get("mode") == "manual"
                                         else "calculés depuis les zones"))
        n = len(self._zones_in(self.curb))
        self.w_detail_hint.value = (f"<small style='color:#777'>{e(b['name'])} : {n} zone(s). La surface sert à "
                                    "pondérer le bilan bâtiment quand il est calculé depuis plusieurs zones."
                                    "</small>")

    # ---------- bâtiments
    def _load_building(self, i):
        self.curb = min(max(i, 0), len(self.buildings) - 1)
        b = self.buildings[self.curb]
        self._loading = True
        try:
            self.w_bld_name.value = b["name"]
            self.w_bld_mode.value = b.get("mode", "auto")
            for k in BUILDING_KEYS:
                self.bld_val[k].value = str(b["results"][k].get("value", "") or "")
                self.bld_max[k].value = str(b["results"][k].get("max", "") or "")
        finally:
            self._loading = False
        self._refresh_building_panel()

    def _refresh_building_panel(self):
        b = self.buildings[self.curb]
        manual = b.get("mode") == "manual"
        self.bld_manual.layout.display = "" if manual else "none"
        zs = [self.zones[i]["name"] for i in self._zones_in(self.curb)]
        if manual:
            txt = ("Saisir les résultats bâtiment de Pléiades (valeur et seuil max). Le DH reste lu "
                   "dans les groupes des zones du bâtiment.")
        elif len(zs) <= 1:
            txt = "Bâtiment à une zone : les résultats bâtiment sont ceux de la zone."
        else:
            txt = ("Plusieurs zones : moyenne pondérée par la surface des zones (valeur et seuil). "
                   "Pour coller exactement aux résultats Pléiades, préférer la saisie directe.")
        zl = ", ".join(_html.escape(z) for z in zs) if zs else "<b style='color:#B3261E'>aucune zone</b>"
        self.bld_info.value = f"<small>{txt}<br>Zones du bâtiment : {zl}</small>"
        self._refresh_crumb()

    def _goto(self, zi):
        """Sélectionne la zone zi (et son bâtiment) : la saisie en cours est sauvegardée avant."""
        self._sync_current()
        self._load_zone(zi)
        self._load_building(int(self.zones[self.cur].get("building", 0)))
        self._refresh_nav()

    def _select_building(self, bi):
        if bi == self.curb:
            return
        zs = self._zones_in(bi)
        if not zs:   # bâtiment sans zone : une zone vide est créée pour pouvoir saisir
            self._sync_current()
            z = new_zone(f"Zone {len(self.zones) + 1}", bi)
            apply_theme_colors({"zones": [z]}, self.w_brand.value)
            self.zones.append(z)
            zs = [len(self.zones) - 1]
            self._mark_dirty()
            self._status("info", f"« {self.buildings[bi]['name']} » n'avait pas de zone : une zone vide a été créée.")
        self._goto(zs[0])

    def _on_building_rename(self, change):
        if self._loading:
            return
        self.buildings[self.curb]["name"] = change["new"].strip() or f"Bâtiment {self.curb + 1}"
        self._refresh_nav()
        self._refresh_crumb()
        self._mark_dirty()

    def _on_building_mode(self, change):
        if self._loading:
            return
        self.buildings[self.curb]["mode"] = change["new"]
        self._refresh_building_panel()
        self._mark_dirty()

    def _on_building_values(self, change):
        if self._loading:
            return
        b = self.buildings[self.curb]
        for k in BUILDING_KEYS:
            b["results"][k] = {"value": self.bld_val[k].value, "max": self.bld_max[k].value}
        self._mark_dirty()

    def building_add(self):
        self._sync_current()
        self.buildings.append(new_building(f"Bâtiment {len(self.buildings) + 1}"))
        bi = len(self.buildings) - 1
        z = new_zone("Zone 1", bi)       # un bâtiment a toujours au moins une zone
        apply_theme_colors({"zones": [z]}, self.w_brand.value)
        self.zones.append(z)
        self._load_zone(len(self.zones) - 1)
        self._load_building(bi)
        self._refresh_nav()
        self._mark_dirty()
        self._status("info", f"« {self.buildings[bi]['name']} » ajouté avec une zone vide : renommez-le, "
                             f"puis saisissez ou importez ses valeurs.")

    def building_delete(self):
        if len(self.buildings) < 2:
            self._status("error", "Impossible de supprimer le dernier bâtiment.")
            return
        self._sync_current()
        bi = self.curb
        name = self.buildings.pop(bi)["name"]
        removed = [z["name"] for z in self.zones if int(z.get("building", 0)) == bi]
        self.zones = [z for z in self.zones if int(z.get("building", 0)) != bi]
        for z in self.zones:
            if int(z.get("building", 0)) > bi:
                z["building"] = int(z["building"]) - 1
        if not self.zones:
            self.zones = [new_zone("Zone 1", 0)]
        self._load_zone(0)
        self._load_building(int(self.zones[0].get("building", 0)))
        self._refresh_nav()
        self._mark_dirty()
        msg = f"Bâtiment « {name} » supprimé"
        self._status("info", msg + (f" avec ses zones : {', '.join(removed)}." if removed else "."))

    # ---------- zones
    def _sync_current(self):
        """Recopie la saisie des onglets dans la zone sélectionnée."""
        z = self.zones[self.cur]
        z["name"] = self.w_zone_name.value.strip() or f"Zone {self.cur + 1}"
        z["indicators"] = {k: t.get_raw() for k, t in self.tabs.items()}

    def _load_zone(self, i):
        self.cur = min(max(i, 0), len(self.zones) - 1)
        z = self.zones[self.cur]
        self._loading = True
        try:
            self.w_zone_name.value = z["name"]
            self.w_zone_surf.value = str(z.get("surface", "") or "")
            for k, t in self.tabs.items():
                t.set_raw(z["indicators"][k])
        finally:
            self._loading = False
        self._refresh_all_feedback()

    def _on_zone_select(self, change):
        if self._loading or change["new"] is None or change["new"] == self.cur:
            return
        self._safe(functools.partial(self._goto, change["new"]))()

    def _on_zone_rename(self, change):
        if self._loading:
            return
        self.zones[self.cur]["name"] = change["new"].strip() or f"Zone {self.cur + 1}"
        self._refresh_nav()
        self._refresh_building_panel()
        self._mark_dirty()

    def _on_zone_building(self, change):
        """Déplace la zone sélectionnée vers un autre bâtiment ; la sélection suit la zone."""
        if self._loading or change["new"] is None:
            return
        old = int(self.zones[self.cur].get("building", 0))
        new = int(change["new"])
        if new == old:
            return
        self._sync_current()
        self.zones[self.cur]["building"] = new
        self._load_building(new)
        self._refresh_nav()
        self._mark_dirty()
        msg = f"Zone « {self.zones[self.cur]['name']} » déplacée vers « {self.buildings[new]['name']} »."
        if not self._zones_in(old):
            msg += f" Attention : « {self.buildings[old]['name']} » n'a plus de zone."
        self._status("info", msg)

    def _on_zone_surface(self, change):
        if self._loading:
            return
        self.zones[self.cur]["surface"] = change["new"]
        self._refresh_crumb()
        self._mark_dirty()

    def _on_brand_change(self, change):
        """Bascule charte WSP / style de base : couleurs par défaut mises à jour, graphiques à actualiser."""
        if self._loading:
            return
        self._sync_current()
        apply_theme_colors({"zones": self.zones}, bool(change["new"]))
        self._load_zone(self.cur)
        self._mark_dirty()

    def zone_add(self):
        self._sync_current()
        z = new_zone(f"Zone {len(self._zones_in(self.curb)) + 1}", self.curb)
        apply_theme_colors({"zones": [z]}, self.w_brand.value)
        self.zones.append(z)
        self._goto(len(self.zones) - 1)
        self._mark_dirty()
        self._status("info", f"Zone « {z['name']} » ajoutée à « {self.buildings[self.curb]['name']} » : "
                             f"renommez-la puis saisissez ses valeurs.")

    def zone_duplicate(self):
        self._sync_current()
        z = copy.deepcopy(self.zones[self.cur])
        z["name"] += " (copie)"
        self.zones.insert(self.cur + 1, z)
        self._load_zone(self.cur + 1)
        self._refresh_building_panel()
        self._refresh_nav()
        self._mark_dirty()
        self._status("info", f"Zone dupliquée : « {z['name']} ».")

    def zone_delete(self):
        idx = self._zones_in(self.curb)
        if len(idx) < 2:
            self._status("error", "Un bâtiment garde au moins une zone : supprimez plutôt le bâtiment.")
            return
        pos = idx.index(self.cur)
        name = self.zones.pop(self.cur)["name"]
        idx = self._zones_in(self.curb)
        self._load_zone(idx[min(pos, len(idx) - 1)])
        self._refresh_building_panel()
        self._refresh_nav()
        self._mark_dirty()
        self._status("info", f"Zone « {name} » supprimée.")

    # ---------- état
    def get_state(self):
        self._sync_current()
        return {"version": 3, "settings": self._settings_raw(),
                "buildings": copy.deepcopy(self.buildings), "zones": copy.deepcopy(self.zones),
                "current_zone": self.cur, "current_building": self.curb}

    def set_state(self, state):
        st = normalize_state(state)
        s, _ = validate_settings(st["settings"])
        self._loading = True
        try:
            self.w_title.value = s.title
            self.w_subtitle.value = s.subtitle
            self.w_footer.value = s.footer
            self.w_width.value = s.width_cm
            self.w_h_over.value = s.height_overview_cm
            self.w_h_single.value = s.height_single_cm
            self.w_font.value = s.font_size
            self.w_show.value = s.show_values
            self.w_hide0.value = s.hide_zero_legend
            self.w_common.value = s.common_scale
            self.w_dec.value = s.decimals
            self.w_bg.value = s.background
            self.w_sep.value = s.decimal_sep
            self.w_brand.value = s.brand
            self.w_dpi.value = s.dpi
            self.w_dir.value = s.export_dir
            self.w_prefix.value = s.file_prefix
            self.w_png.value, self.w_svg.value, self.w_pdf.value = (f in s.formats for f in ("png", "svg", "pdf"))
        finally:
            self._loading = False
        self.buildings = st["buildings"]
        self.zones = apply_theme_colors(st, s.brand)["zones"]
        self.cur = min(max(st.get("current_zone", 0), 0), len(self.zones) - 1)
        self._load_zone(self.cur)
        self._load_building(int(self.zones[self.cur].get("building", 0)))   # le bâtiment suit la zone
        self._refresh_nav(force=True)
        self._cache = {}
        self._data_dirty = True
        self._show_previews()

    def _on_tab_change(self, tab):
        if self._loading:
            return
        s, _ = validate_settings(self._settings_raw())
        self._update_tab_title(tab, tab.refresh_feedback(s))
        self._sync_current()
        self._refresh_nav()
        self._mark_dirty()

    def _update_tab_title(self, tab, res):
        i = INDICATOR_KEYS.index(tab.key)
        flag = "" if (res.ok or not tab.include.value) else " ⚠"
        excl = " (exclu)" if not tab.include.value else ""
        self.tab_widget.set_title(i, f"{tab.key}{flag}{excl}")

    def _refresh_all_feedback(self):
        if self._loading:
            return
        s, _ = validate_settings(self._settings_raw())
        for t in self.tabs.values():
            self._update_tab_title(t, t.refresh_feedback(s))

    def _collect(self):
        """Valide toutes les zones et calcule les bilans bâtiment.
        Retourne (settings, [(zone, {indicateur: résultat})], erreurs, avertissements, bilans, projet conforme)."""
        self._sync_current()
        s, warns = validate_settings(self._settings_raw())
        self._refresh_all_feedback()
        self._refresh_nav()
        multi = len(self.zones) > 1
        multi_b = len(self.buildings) > 1
        zones_valid, errors = [], []
        for z in self.zones:
            disp = f"{self._bname(z)} — {z['name']}" if multi_b else z["name"]
            results = {}
            for k in INDICATOR_KEYS:
                raw = z["indicators"][k]
                if raw["include"]:
                    results[k] = validate_indicator(k, raw)
                    for e in results[k].errors:
                        errors.append((f"{disp} / " if multi else "") + f"{k} — {e}")
            zones_valid.append((disp, results))
        if not any(zr for _, zr in zones_valid):
            warns.append("Aucun indicateur inclus.")
        bilans, project_ok = project_bilan({"buildings": self.buildings, "zones": self.zones})
        return s, zones_valid, errors, warns, bilans, project_ok

    @staticmethod
    def _bilan_notes(bilans, project_ok):
        notes = []
        if project_ok:
            notes.append("Bilan : tous les indicateurs sont sous leur seuil → « Le projet est conforme à la RE2020 ».")
            return notes
        for b in bilans:
            for e in b.errors:
                notes.append(f"Bilan {b.name} : {e}")
            for w in b.warnings:
                notes.append(f"Bilan {b.name} : {w}")
            for it in b.exceed:
                notes.append(f"Bilan {b.name} : {it.label} dépasse le seuil max "
                             f"({fmt_gap(it.gap, ',')}) — mention de conformité non affichée.")
        return notes

    # ---------- familles de graphiques et actualisation
    def _checked(self):
        return [k for k in FAMILY_KEYS if self.chips[k].value]

    def _mark_dirty(self):
        """Une saisie ou un réglage a changé : les aperçus existants ne sont plus à jour."""
        if self._loading:
            return
        self._data_dirty = True
        self._refresh_banner()

    def _update_chips(self):
        for k, lab, _ in FAMILIES:
            n = self._counts.get(k)
            t = self.chips[k]
            t.description = lab if n is None else f"{lab} · {n} fig."
            t.icon = "check" if t.value else ""

    def _update_counts(self):
        """Compte les figures de chaque famille (planification seule, sans tracé)."""
        s, zones_valid, _, _, bilans, project_ok = self._collect()
        views = plan_family_views(s, zones_valid, bilans, project_ok, FAMILY_KEYS)
        self._counts = {k: 0 for k in FAMILY_KEYS}
        self._counts.update(Counter(view_family(v) for v in views))
        self._update_chips()
        self._refresh_banner()

    def _refresh_banner(self):
        fams = self._checked()
        if not fams:
            self.w_banner.value = _banner_html("info", "Aucune famille cochée : cochez au moins un type de graphique.")
            self.btn_refresh.disabled = True
            return
        self.btn_refresh.disabled = False
        todo = fams if self._data_dirty else [f for f in fams if f not in self._cache]
        if todo:
            n = [self._counts.get(f) for f in todo]
            nb = f" · {sum(n)} figure(s) à calculer" if all(x is not None for x in n) else ""
            head = "Modifications non appliquées" if self._data_dirty and self._cache else "Graphiques à calculer"
            self.w_banner.value = _banner_html("warning", f"{head}{nb} — cliquer sur « Actualiser ».")
            self.btn_refresh.button_style = "primary"
        else:
            n = sum(len(self._cache.get(f, [])) for f in fams)
            self.w_banner.value = _banner_html("success", f"✓ À jour · {n} figure(s)")
            self.btn_refresh.button_style = ""

    def _on_family(self, change):
        if self._loading:
            return
        self._update_chips()
        self._show_previews()      # les familles déjà calculées s'affichent sans recalcul
        self._refresh_banner()

    def _preset(self, fams):
        self._loading = True
        try:
            for k, t in self.chips.items():
                t.value = k in fams
        finally:
            self._loading = False
        self._update_chips()
        self._show_previews()
        self._refresh_banner()

    def _image_entry(self, caption, fig):
        data = figure_bytes(fig, fmt="png", dpi=PREVIEW_DPI, transparent=False)
        return caption, data, int(fig.get_size_inches()[0] * PREVIEW_DPI)

    def generate(self):
        """Calcule les aperçus des familles cochées qui ne sont pas encore à jour."""
        s, zones_valid, errors, notes, bilans, project_ok = self._collect()
        all_views = plan_family_views(s, zones_valid, bilans, project_ok, FAMILY_KEYS)
        self._counts = {k: 0 for k in FAMILY_KEYS}
        self._counts.update(Counter(view_family(v) for v in all_views))
        if self._data_dirty:
            self._cache = {}
        fams = self._checked()
        todo = [f for f in fams if f not in self._cache]
        built, noted = 0, set()
        for f in todo:
            self._cache[f] = []
        for v in all_views:
            f = view_family(v)
            if f not in todo:
                continue
            fig, note = build_view(v, s)
            if note and f not in noted:
                notes.append(f"{FAMILY_LABEL[f]} : {note}.")
                noted.add(f)
            self._cache[f].append(self._image_entry(v["caption"], fig))
            del fig
            built += 1
        self._data_dirty = False
        self._update_chips()
        self._show_previews()
        blocked = [n for n, r in zones_valid if r and any(not x.ok for x in r.values())]
        if blocked and set(fams) & {"zone", "single"}:
            notes.append("Vue d'ensemble non générée pour : " + ", ".join(blocked) + ".")
        notes = self._bilan_notes(bilans, project_ok) + notes
        stamp = f"{now_local():%H:%M:%S}"
        shown = sum(len(self._cache.get(f, [])) for f in fams)
        head = (f"Graphiques actualisés à {stamp} : {shown} figure(s) affichée(s)"
                + (f", {built} calculée(s)." if built != shown else "."))
        if not fams:
            self._status("info", "Aucune famille cochée à l'étape 3 : rien à calculer.", notes)
        elif errors:
            self._status("error", head + " Des saisies sont à corriger (étape 2).", errors + notes)
        else:
            self._status("success" if project_ok else "warning", head, notes)
        self._refresh_banner()
        return not errors

    # ---------- aperçu (un graphique à la fois)
    def _show_previews(self):
        fams = self._checked()
        items = [it for f in fams for it in self._cache.get(f, [])]
        old_cap = None
        if self._previews and isinstance(self.w_prev_sel.value, int) and self.w_prev_sel.value < len(self._previews):
            old_cap = self._previews[self.w_prev_sel.value][0]
        self._previews = items
        caps = [c for c, _, _ in items]
        idx = caps.index(old_cap) if old_cap in caps else 0
        self._loading_prev = True
        try:
            self.w_prev_sel.options = [(c, i) for i, c in enumerate(caps)]
            self.w_prev_sel.value = idx if items else None
        finally:
            self._loading_prev = False
        if items:
            self._show_preview(idx)
            return
        self.w_img.layout.display = "none"
        self.w_prev_pos.value = ""
        if not fams:
            msg = _msg_html("info", "Cochez au moins une famille de graphiques.")
        elif any(f in self._cache for f in fams):
            hint = []
            if "ind" in fams and len(self.zones) < 2:
                hint.append("Indicateurs : au moins deux zones sont nécessaires.")
            if "bld" in fams and len(self.buildings) < 2:
                hint.append("Bâtiment(s) : un seul bâtiment, son bilan est dans « Synthèse ».")
            msg = _msg_html("info", "Aucune figure pour les familles cochées.", hint)
        else:
            msg = _msg_html("info", "Cliquer sur « Actualiser » pour calculer les graphiques cochés.")
        self.w_prev_msg.value = msg

    def _show_preview(self, i):
        if not self._previews:
            return
        i = min(max(int(i), 0), len(self._previews) - 1)
        cap, data, w_px = self._previews[i]
        self.w_img.value = data
        self.w_img.layout.width = f"{w_px}px"
        self.w_img.layout.display = ""
        self.w_prev_msg.value = ""
        self.w_prev_pos.value = f"<span style='color:#777'>&nbsp;{i + 1} / {len(self._previews)}</span>"

    def _on_prev_select(self, change):
        if self._loading_prev or change["new"] is None:
            return
        self._show_preview(change["new"])

    def _step_preview(self, d):
        n = len(self._previews)
        if not n:
            return
        cur = self.w_prev_sel.value if isinstance(self.w_prev_sel.value, int) else 0
        self.w_prev_sel.value = (cur + d) % n

    # ---------- export (calcul à la demande des familles demandées)
    def _offer_download(self, data, filename, label):
        size = len(data) / 1e6
        self.dl_box.value = (download_link(data, filename, f"{label} ({size:.1f} Mo)")
                             + "<br><small>Si rien ne se passe au clic : clic droit sur le bouton → "
                               "« Enregistrer le lien sous… ».</small>")

    def _export(self, fams):
        s, zones_valid, errors, notes, bilans, project_ok = self._collect()
        if not s.formats:
            self._status("error", "Aucun format d'export sélectionné (PNG, SVG ou PDF).")
            return
        if not fams:
            self._status("error", "Aucune famille cochée à l'étape 3 : choisissez quoi exporter.")
            return
        views = plan_family_views(s, zones_valid, bilans, project_ok, fams)
        if "ind" in fams and len(zones_valid) < 2:
            notes.append("Indicateurs : au moins deux zones sont nécessaires.")
        files, paths = [], []
        for v in views:
            fig, note = build_view(v, s)
            if note:
                notes.append(f"{v['caption']} : {note}.")
            for fmt in s.formats:
                files.append((f"{v['basename']}.{fmt}",
                              figure_bytes(fig, fmt=fmt, dpi=s.dpi, transparent=s.transparent)))
            if self.w_disk.value:
                paths += export_figure(fig, v["basename"], s)
            del fig
        if not files:
            self.dl_box.value = ""
            self._status("error", "Rien n'a été exporté.", errors + notes)
            return
        stamp = f"{now_local():%Y%m%d_%H%M}"
        name = f"{safe_filename(s.file_prefix)}_graphiques_{stamp}.zip"
        self._offer_download(figures_zip_bytes(files), name, f"Télécharger {name}")
        items = [f"{len(views)} graphique(s), {len(files)} fichier(s) dans le ZIP ("
                 + ", ".join(FAMILY_LABEL[f] for f in FAMILY_KEYS if f in fams) + ")."]
        items += [f"Écrit sur le disque : {p}" for p in paths] + errors + notes
        kind = "warning" if errors else "success"
        head = "Export partiel (éléments à corriger ignorés)" if errors else "Export prêt"
        self._status(kind, f"{head} : cliquez sur le bouton de téléchargement ci-dessous "
                           f"(fond {s.background}, PNG {s.dpi} dpi).", items)

    def export_selected(self):
        what = self.w_exp_what.value
        if what == "checked":
            fams = set(self._checked())
        elif what == "all":
            fams = set(FAMILY_KEYS)
        else:
            fams = {what}
        self._export(fams)

    def export_all(self):
        self._export(set(FAMILY_KEYS))

    # ---------- Excel
    def import_excel_file(self):
        if not self.w_excel_upload.value:
            self._status("error", "Aucun fichier Excel sélectionné.")
            return
        v = self.w_excel_upload.value
        uploaded = next(iter(v)) if isinstance(v, (tuple, list)) else next(iter(v.values()))
        data = uploaded["content"] if isinstance(uploaded, dict) else uploaded.content
        self.w_paste.value = excel_upload_to_text(bytes(data))
        self.paste_import()

    def paste_import(self):
        if not self.w_paste.value.strip():
            self._status("error", "La zone de collage est vide.")
            return
        try:
            state, rep = apply_pasted_table(self.get_state(), self.w_paste.value, self.w_paste_zone.value)
        except ValueError as e:
            self._status("error", "Collage non interprété.", [str(e)])
            return
        state["current_zone"] = min(self.cur, len(state["zones"]) - 1)
        self.set_state(state)
        ok = self.generate()
        items = [f"Mis à jour — {x}" for x in rep["updated"]]
        items += [f"Créé — {x}" for x in rep["created"]]
        items += rep["warnings"]
        if rep["ignored"]:
            items.append("Clés non gérées par ce notebook (ignorées) : " + ", ".join(rep["ignored"]))
        if rep["unknown"]:
            items.append("Lignes non reconnues (ignorées) : " + ", ".join(rep["unknown"][:12])
                         + (" …" if len(rep["unknown"]) > 12 else ""))
        if not ok:
            items.append("Certaines saisies sont à corriger : voir « Saisies à corriger » à l'étape 2.")
        kind = "success" if ok and not rep["warnings"] and not rep["unknown"] else "warning"
        self._status(kind, "Valeurs importées, graphiques cochés actualisés.", items)

    def paste_clear(self):
        self.w_paste.value = ""

    def to_excel(self):
        st = self.get_state()
        text, warns = state_to_excel_text(st, self.w_sep.value)
        self.w_excel_out.value = text
        self._status("warning" if warns else "success",
                     "Tableau généré : copiez-le (Ctrl+A, Ctrl+C) et collez-le en A1 dans Excel.", warns)

    # ---------- projet
    def save(self):
        st = self.get_state()
        name = os.path.basename(self.w_json.value.strip() or "projet_RE2020.json")
        if not name.lower().endswith(".json"):
            name += ".json"
        self._offer_download(project_json_bytes(st), name, f"Télécharger {name}")
        items = []
        if self.w_disk.value:
            items.append(f"Écrit sur le disque : {save_project_json(st, self.w_json.value)}")
        self._status("success", "Projet prêt : cliquez sur le bouton de téléchargement ci-dessous.", items)

    def load_upload(self):
        v = self.w_json_upload.value
        if not v:
            self._status("error", "Aucun fichier JSON choisi (bouton « Choisir un JSON »).")
            return
        up = next(iter(v)) if isinstance(v, (tuple, list)) else next(iter(v.values()))
        data = up["content"] if isinstance(up, dict) else up.content
        name = up.get("name", "") if isinstance(up, dict) else getattr(up, "name", "")
        try:
            state = project_from_bytes(data)
        except ValueError as e:
            self._status("error", "Chargement impossible.", [str(e)])
            return
        self.set_state(state)
        self.generate()
        self._status("success", "Projet chargé, graphiques cochés actualisés.", [name] if name else [])

    def load(self):
        try:
            state, path = load_project_json(self.w_json.value)
        except (OSError, ValueError) as e:
            self._status("error", "Chargement impossible.",
                         [str(e), "En ligne, utilisez « Choisir un JSON » puis « Ouvrir »."])
            return
        self.set_state(state)
        self.generate()
        self._status("success", "Projet chargé, graphiques cochés actualisés.", [path])

    def load_demo(self):
        self.set_state(demo_state())
        self.generate()
        self._status("info", "Projet exemple chargé : 2 bâtiments, zones tertiaire et commerce, "
                             "valeurs FICTIVES de démonstration.")

    def new_project(self):
        self.set_state(default_state())
        self.dl_box.value = ""
        self._counts = {}
        self._update_chips()
        self._status("info", "Projet vierge : décrivez les bâtiments et zones, importez ou saisissez les "
                             "valeurs (étape 2), puis « Actualiser » à l'étape 3.")

    def show(self):
        display(self.widget)
```
